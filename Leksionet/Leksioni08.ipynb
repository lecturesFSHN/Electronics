# **Fizika e gjysmëpërçuesve**

## **1. Struktura atomike dhe kristalore e gjysmëpërçuesve**

### 1.1 Struktura atomike – Gjysmëpërçuesit si elementë të grupit IV
Gjysmëpërçuesit klasikë si **silici (Si)** dhe **germaniumi (Ge)** ndodhen në **grupin IV të tabelës periodike**. Kjo do të thotë që kanë **katër elektrone në shtresën valencë**.

Në një kuptim mekaniko-kuantik:
- Elektronet rrethojnë bërthamën në **nivele energjie** të lejueshme, që përshkruhen nga zgjidhjet e ekuacionit të Schrödinger për potencialin elektrostatik bërthamë-elektron.
- Në atom të vetëm, këto nivele janë **diskrete** (energjitë janë të kuantizuara).

#### Për shembull për Silicin:
- Konfigurimi elektronik është:  
  $$
  \text{Si: } 1s^2 2s^2 2p^6 3s^2 3p^2
  $$
- Katër elektronet në nivelin $n=3$ janë ata që marrin pjesë në lidhje kimike – **elektronet e valencës**.

> Këta elektrone janë gjithashtu përgjegjës për **sjelljen elektrike** të materialit.



### 1.2 Lidhja kovalente dhe rrjeti kristalor

Në fazën e ngurtë, atomet e silicit organizohen në një **strukturë kristaline** të ngjashme me **strukturën kristaline të diamantit**.

#### Struktura kristaline:
- Çdo atom është i lidhur me **katër fqinj të afërt** përmes **lidhjeve kovalente**.
- Atomet ndajnë mes njëri-tjetrit çifte elektronesh që formojnë lidhje te forta dhe një **rrjetë shumë të qëndrueshëm kristali**.
- Në këtë gjendje ideale, elektronet janë të **lokalizuara në lidhje** dhe nuk janë të lirë të kontribuojnë në përçueshmëri.

> Kjo shpjegon pse një kristal Silici i pastër në temperaturë shumë të ulët sillet si një **izolator**.

Qeliza elementare e këtij kristali (njësia themelore përsëritëse në hapësirë) ka simetri **kubike faqe centruar (FCC)**, shiko figurën.

<img src="../figs/lec6/Si_crystal.svg" alt="Si" width="400" style="background-color:#fff;" >



### 1.3 Nga nivele diskrete në breza energjie

Në një atom të vetëm, elektronet lejohen vetëm në disa nivele energjie specifike.

Nga ana tjetër, në një **kristal me shumë atome të afërta**, orbitalet atomike të valencës **mbivendosen** dhe ndërveprojnë.

#### Interferenca e niveleve:
- Kur $N \sim 10^{23}$ atome identike ndërveprojnë, nivelet diskrete të energjisë mbivendosen në $N$ nivele shumë të afërta.
- Këto nivele formojnë një **brez (bandë) energjie të vazhdueshëm**.

Rezultati:
- Formohet një **brez valencë** (i mbushur me elektrone)
- Dhe një **brez përçueshmërie** (zakonisht bosh)

> Elektronet mund të kontribuojnë në rrymën elektrike vetëm nëse ndodhen në brezin (bandën) e përçueshmërisë.


<img src="../figs/lec6/Si_bands.svg" alt="Si" width="700"  style="background-color:#fff;" >


### 1.4 Energjia potenciale periodike në kristal

Energjia që ndjen një elektron në një kristal është **periodike**:
$$ V(x) = V(x + a), $$
ku $a$ është perioda hapsinore (distancat mes nyjeve atomike)

#### Zgjidhja kuantike:
- Ekuacioni i Schrödinger në këtë potencial jep **zgjidhje të formës së Bloch**:
$$ \psi_k(x) = u_k(x) e^{ikx} $$

- Kjo strukturë krijon:
  - **breza të lejuar energjie**
  - **zona të ndaluara (band gaps)**


#### **Brezi i valencës dhe brezi i përçueshmërisë**

Pas formimit të brezave, kemi dy breza të rëndësishëm:

#### Brezi i valencës (*Valence band*):
- Brezi që rezulton nga ndërveprimi i **orbitalëve të valencës**.
- Në temperaturë të ulët, ky brez është **i mbushur plotësisht** me elektrone.

#### Brezi i përçueshmërisë (*Conduction band*):
- Brezi më i lartë energjetik.
- Në gjysmëpërçues, ky brez është **i zbrazët në $T=0 K$**, por mund të popullohet me elektrone që fitojnë energji (psh në $T>0$).

#### Hapësira ndërmjet dy brezave – **band gap $E_g$**:
$$ E_g = E_c - E_v $$


<img src="../figs/lec6/material_bands.svg" alt="Si" width="700"  style="background-color:#fff;" >


### Parimet bazë të përçueshmërinë

**Një brez i mbushur plotësisht nuk mund të përçojë rrymë.**
- Elektronet kanë nevojë për **gjendje të lira afër në energji** për të lëvizur.
- Kjo ndodh vetëm në:
  - breza pjesërisht të mbushur  
  - ose në brezin e përçueshmërisë


### **Shembuj konkretë të band gap-it**

| Materiali       | Lloji          | Band Gap (eV) |
|----------------|----------------|----------------|
| Germanium (Ge) | Gjysmëpërçues  | 0.66           |
| Silici (Si)  | Gjysmëpërçues  | 1.12           |
| Arsenid Galliumi (GaAs) | Gjysmëpërçues | 1.42 |
| Karboni (diamant) | Izolator     | ∼5.5           |
| Bakër (Cu)     | Përçues        | 0 (banda mbivendosen) |

#### Vëzhgim:
- Sa më i vogël band gap-i, aq më **lehtë elektronët kalojnë në brez përçueshmërie**.

Në temperaturë dhome:
$$
k_B T \approx 0.025 \text{ eV}
$$

> Kjo energji është e mjaftueshme që një pjesë e vogël e elektroneve të eksitohet në brezin e përçueshmërisë.



### Lidhja me përçueshmërinë

#### Gjysmëpërçuesit:

Kur një elektron kalon në brezin e përçueshmërisë:
- ai bëhet një **bartës negativ (elektron i lirë)**

Por në brezin e valencës mbetet një mungesë elektroni:

Kjo mungesë quhet **vrimë (hole)**
- Vrima sillet si një **bartës pozitiv**
- Lëvizja e elektroneve në lidhjet kovalente mund të interpretohet si **lëvizje e vrimave në drejtim të kundërt**

> Pra krijohet një çift:
> - elektron (në conduction band)
> - vrimë (në valence band)


#### Mekanizmi i përçueshmërisë

>  Rryma elektrike në gjysmëpërçues është rezultat i:
> - lëvizjes së **elektroneve në brezin e përçueshmërisë**
> - dhe lëvizjes së **vrimave në brezin e valencës**


---


Me pak fjalë, sjellja elektrike e materialit varet nga:
- **madhësia e band gap-it**
- dhe nga fakti nëse elektronet mund të kalojnë nga brezi i valencës në atë të përçueshmërisë.

Këto karakteristika i klasifikojnë materialet në:
- **Përçues (Metalet)**: nuk ka band gap → elektronet lëvizin lirshëm.
  * Elektronet janë shumë të lirë → lëvizin lehtësisht në fushë elektrike.
  * Përçueshmëri shumë e lartë (∼10⁶ – 10⁸ S/m).
- **Izolatorë**: band gap shumë i madh → nuk ka lëvizje (nuk ka elektrone të lira)
   * Band gap shumë i madh → nuk ka bartës të lirë.
   * Përçueshmëri shumë e ulët.
- **Gjysmëpërçues**: band gap i vogël → disa elektrone mund të eksitohen termikisht dhe të kalojnë në bandën e përçueshmërisë ku janë të lirë të levizin, si dhe roli i vrimave.
  * Band gap i vogël → krijohen disa bartës termikë.
  * Përçueshmëria është e vogël, por **e kontrollueshme**.

> Kjo krijon **bartës ngarkese (carriers)** që mund të transportojnë rrymë.


<img src="../figs/lec7/conductance.png" alt="Vdiode" width="600" style="background-color:#fff;" >



### Përmbledhje:

- Struktura atomike dhe kristalore përcakton **strukturën e brezave energjetikë**.
- Brezat përcaktojnë nëse ekzistojnë **elektrone të lëvizshme**.
- Band gap-i kontrollon **numrin e bartësve të ngarkesës**
- Bartësit në gjysmëpërçues janë:
  - elektronet  
  - vrimat  

Në vijim:
> Do të shohim si mund të **kontrollojmë numrin dhe llojin e këtyre bartësve** (përmes dopimit), dhe si krijohen struktura ku përçueshmëria ndryshon në hapësirë — si p–n junction.

## **2. Bartësit e ngarkesës dhe dopimi në gjysmëpërçues**

### **2.1 Përçueshmëria në gjysmëpërçues të pastër (intrinsik)**

Në një gjysmëpërçues të pastër, si **silici i pastër**, në temperaturë të ulët (0 K) brezi i përçueshmërisë është i zbrazët, ndërsa brezi i valencës është i mbushur.

Por me **rritjen e temperaturës**, disa elektrone marrin **energji termike** $(k_B T \sim 0.025 \text{ eV}$ në 300 K) dhe **kalojnë në brezin e përçueshmërisë**.

Kjo sjell:
1. **Elektron i lirë** në conduction band  
2. **Vrima** në valence band  

$$ \text{e}^- \quad \leftrightarrow \quad \text{vrimë}^+ $$

> Këta janë **bartësit e ngarkesës** që mund të kontribuojnë në rrymë.

#### Mekanizmi i lëvizjes:
- **Drift**: lëvizje nën ndikimin e një fushe elektrike  
- **Diffusion**: lëvizje nga zona me përqendrim të lartë drejt atij të ulët  

> Këto dy mekanizma janë themelorë për sjelljen e pajisjeve gjysmëpërçuese.



### **2.2 Dopimi – Papastërtia e kontrolluar për të rritur përçueshmërinë**

Për të rritur përçueshmërinë, shtojmë **papastërti të kontrolluara** (~1 në 10⁷ atome).

Materiali bëhet **ekstrinsik** dhe numri i bartësve rritet ndjeshëm.

<img src="../figs/lec6/Si_doping.svg" width="900" style="background-color:#fff;" > 


### **2.3 Dopimi i tipit $n$**

- Elemente të grupit V-A (P, As)  
- Jepin **një elektron të lirë**

$$ \text{P}^0 \rightarrow \text{P}^+ + e^- $$

- Elektronet janë **bartës shumicë**

> **Niveli i Fermit zhvendoset drejt brezit të përçueshmërisë**



### **2.4 Dopimi i tipit $p$**

- Elemente të grupit III-A (B, Al)  
- Krijojnë **vrima**

$$ \text{B}^0 + e^- \rightarrow \text{B}^- + \text{vrimë}^+ $$

- Vrimat janë **bartës shumicë**

> **Niveli i Fermit zhvendoset drejt brezit të valencës**



### **2.5 Përçueshmëria totale**

$$ \sigma = q (n \mu_n + p \mu_p) $$

- $n, p$ → përqendrimet  
- $\mu_n, \mu_p$ → mobilitetet  

> Përçueshmëria varet si nga **numri i bartësve**, ashtu edhe nga **aftësia e tyre për të lëvizur**.


### **2.6 Drejt bashkimeve p–n**

Tani kemi dy lloje materialesh:

- rajone **n-tip** (shumë elektrone)
- rajone **p-tip** (shumë vrima)

Kur këto dy rajone vendosen në kontakt:
- elektronet difuzojnë nga n → p  
- vrimat difuzojnë nga p → n  

Kjo krijon:
- ndarje ngarkese  
- një fushë elektrike të brendshme  

> Ky është hapi i parë drejt formimit të një **p–n junction** (kalimi p–n), që do të analizojmë në vijim.


<img src="../figs/lec6/Pn-junction-equilibrium.svg" width="900" style="background-color:#fff;" > 


### **2.7 Formimi i zonës së varfëruar (depletion region)**

Kur bartësit difuzojnë:

- elektronet që kalojnë në rajonin p **rikombinohen me vrimat**  
- vrimat që kalojnë në rajonin n bëjnë të njëjtën gjë  

Në zonën pranë kufirit:
- nuk mbeten bartës të lirë  
- mbeten vetëm jonet fikse (dopantët e jonizuar)


Kjo krijon:
- një **zonë të varfëruar nga bartësit**
- një **ndarje ngarkese**


### **2.8 Formimi i fushës elektrike të brendshme**

Ngarkesat fikse krijojnë një **fushë elektrike të brendshme**:

- kjo fushë kundërshton difuzionin  
- sistemi arrin një **ekuilibër**

> Ky quhet **potencial i ndërtuar nga ekuilibri (built-in potential)**




### Përmbledhje

- Dopimi kontrollon **llojin dhe numrin e bartësve**
- Transporti ndodh përmes:
  - drift  
  - diffusion
-  Kur këto rajone bashkohen:
  - ndodh difuzion  
  - krijohet një fushë elektrike  
  - formohet një **p–n junction**


Në leksionin vijues:
> Do të shohim se si kjo strukturë çon në një **sjellje shumë jo-lineare të rrymës** dhe si përshkruhet matematikisht (dioda).

## **3. Bashkimi p-n**

### **3.1 Çfarë është një bashkim p-n?**

Kur vendosim një gjysmëpërçues të **tipit $p$** dhe një tjetër të **tipit $n$** pranë njëri-tjetrit, formohet një **bashkim p-n**.

> Në këtë ndërfaqe ndodhin fenomene kryesisht **elektrostatike dhe statistikore të bartësve**, që përcaktojnë sjelljen e pajisjes.



### **3.2 Çfarë ndodh në kontaktin p-n?**

<img src="../figs/lec6/diode.svg" alt="Si" width="500" style="background-color:#fff;" >

#### a) Difuzioni i bartësve:
- Elektronet (nga zona n) difuzojnë drejt zonës p  
- Vrimat (nga zona p) difuzojnë drejt zonës n  

> shkaku: **gradient përqendrimi**



#### b) Rikombinimi:
- Elektronet dhe vrimat rikombinohen → zhduken si bartës të lirë  
- Formohet një zonë pa bartës → **zona e varfëruar (depletion region)**



#### c) Formimi i fushës elektrike:
- Mbeten jone fikse:
  - negative në anën p  
  - pozitive në anën n  

krijohet një **fushë elektrike e brendshme**
- kjo fushë kundërshton difuzionin  
- sistemi arrin ekuilibër



### **3.3 Barrierë potenciali (built-in potential)**

Fusha elektrike krijon një **barrierë energjie** që pengon bartësit:

$$
V_0 \approx \frac{kT}{q} \ln\left( \frac{N_a N_d}{n_i^2} \right)
$$

- Varet nga dopimi dhe temperatura  
- Për silicium: ~0.6–0.7 V  

> Kjo është arsyeja pse bartësit nuk kalojnë lirshëm në ekuilibër.



## **4. Polarizimi i bashkimit p-n**

### **4.1 Polarizimi përpara (forward bias)**
- Fusha e jashtme **zvogëlon barrierën**
- Zona e varfëruar **ngushtohet**

Bartësit mund të kalojnë më lehtë kufirin:
- elektronet nga n → p  
- vrimat nga p → n  

> Rryma rritet shumë shpejt me tensionin



### **4.2 Polarizimi kundërt (reverse bias)**
- Fusha e jashtme **rrit barrierën**
- Zona e varfëruar **zgjerohet**

> Bartësit shumicë bllokohen  
> Kalon vetëm një rrymë shumë e vogël (bartës minoritarë)



### **4.3 Prishja (breakdown)**

Në tension të lartë të kundërt:
- bartësit përshpejtohen fort  
- krijojnë bartës të rinj (avalanche)  
- ose tunelojnë përmes barrierës (Zener)

> Rryma rritet shumë shpejt



### Përmbledhje

- Difuzioni krijon një zonë pa bartës  
- Kjo krijon një fushë elektrike dhe një barrierë  
- Polarizimi ndryshon këtë barrierë  

Kjo strukturë çon në një sjellje shumë jo-lineare të rrymës,  
që do të përshkruajmë matematikisht në leksionin vijues (dioda).

```python
import numpy as np
import matplotlib.pyplot as plt

V = np.linspace(-1, 1, 400)
q = 1.6e-19  # C
k = 1.38e-23  # J/K
T = 300  # K
n = 1.5
Is = 1e-12

I = Is * (np.exp((q * V) / (n * k * T)) - 1)

plt.figure(figsize=(8, 4))
plt.plot(V, I * 1e6)  # mikroampera
plt.axhline(0, color='gray', linewidth=0.5)
plt.axvline(0, color='gray', linewidth=0.5)
plt.title("Karakteristika I-V e një diode gjysmëpërçuese")
plt.xlabel("Tensioni V (V)")
plt.ylabel("Rryma I (μA)")
plt.grid(True)
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt

Vin = np.linspace(-1, 1.5, 500)
R = 1000  # Ohm
Vd_th = 0.7
I_out = np.where(Vin > Vd_th, (Vin - Vd_th) / R, 0)

plt.figure(figsize=(8, 4))
plt.plot(Vin, I_out * 1e3)  # mA
plt.axhline(0, color='gray', linewidth=0.5)
plt.axvline(0, color='gray', linewidth=0.5)
plt.title("Rryma në një qark me diodë dhe rezistencë")
plt.xlabel("Tensioni hyrës Vin (V)")
plt.ylabel("Rryma I (mA)")
plt.grid(True)
plt.show()
```

```python
import ipywidgets as widgets
from IPython.display import display

@widgets.interact(R=widgets.FloatSlider(value=1000, min=100, max=5000, step=100, description='Rezistenca (Ω)'),
                  Vd_th=widgets.FloatSlider(value=0.7, min=0.2, max=1.0, step=0.05, description='Pragu i diodës (V)'))
def interactive_diode(R, Vd_th):
    Vin = np.linspace(-1, 1.5, 500)
    I_out = np.where(Vin > Vd_th, (Vin - Vd_th) / R, 0)
    plt.figure(figsize=(8, 4))
    plt.plot(Vin, I_out * 1e3)
    plt.axhline(0, color='gray', linewidth=0.5)
    plt.axvline(0, color='gray', linewidth=0.5)
    plt.title("Rryma në varësi të Vin për parametra të ndryshëm")
    plt.xlabel("Tensioni hyrës Vin (V)")
    plt.ylabel("Rryma I (mA)")
    plt.grid(True)
    plt.show()
```
