#### **Ideja kryesore e këtij leksioni**

Transistori është një pajisje me tre terminale ku një sinjal i vogël kontrolli mund të komandojë një rrymë më të madhe.

Në këtë leksion fokusohemi te **BJT**:

- terminali i kontrollit është **baza**;
- madhësia kryesore e kontrollit është $V_{BE}$ ose $I_B$;
- rryma kryesore e kontrolluar është $I_C$;
- përdorimet bazë janë **amplifikimi** dhe **komutimi**.

</div>


# **I. Hyrje**

## **Çfarë është një transistor?**

Një transistor është një pajisje elektronike gjysmëpërçuese e cila sillet ose si një çelës i komanduar elektrikisht ose si një amplifikator i kontrolluar elektrikisht. Në thelb të funksionit të transistorit është kontrolli i rrjedhjes së rrymës, analoge me kontrollin e rrjedhjes së ujit nga një rubinet. Ndryshe nga pajisjet pasive që kemi parë deri tani, të cilat kanë pasur vetëm dy terminale, transistorët kanë edhe një terminal shtesë për kontrollin, pra janë pajisje me tre terminale. Një sinjal i vogël tensioni apo rryme që aplikohet mbi terminalin e kontrollit, mund të shkaktojë një ndryshim të madh rryme mbi dy terminalet e tjera të tij. Për shkak të numrit të madh të aplikimeve që ato kanë, transistorët gjenden praktikisht në çdo qark elektrik të pajisjeve moderne.


<div style="border:5px solid #0a0; background-color:#ca0; color:#003; border-radius: 15px; padding: 10px;">

#### **Përkufizimi**

**Një transistor është një pajisje elektronike gjysmëpërçuese me tre terminale që kontrollon rrjedhën e rrymës në një qark, nëpërmjet një sinjali kontrollues të jashtëm (rrymë ose tension).**

- Nëse kontrolli bëhet me rrymë (injeksion bartësish) ➔ kemi transistor bipolar (Bipolar Junction Transistor, BJT shkurt).
- Nëse kontrolli bëhet me tension (kontroll elektrostatik të kanaleve) ➔ kemi transistor me efekt fushe (Field Effect Transistor, FET shkurt).

</div>

- Transistori kontrollon rrymën me një sinjal të vogël elekrik (rrymë apo tension) dhe e **amplifikon** efektin e këtij sinjali në një sinjali shumë më të madh në dalje. Pra **transistori është një amplifikator**.
- Për sinjale mjaftueshmërisht të mëdha, **transistori kalon në regjim ngopje** dhe **kthehet efektivisht në një çelës elektrik të komandueshëm**.
- Ai është **zemra** e elektronikës moderne: mikroprocesorët, sensorët, komunikimi digjital, etj — të gjitha varen nga transistorët.


<br>

**Në thelb**, një transistor është një **valvul elektrike**: përdor një **sinjal të vogël** për të kontrolluar **një sinjal më të madh**.

## Përmbledhje historike e transistorit

- **1925:** Lilienfeld imagjinoi një pajisje qe mund ti komandohej rezistenca elektrike nëpërmjet fushës elektrike, por materialet nuk ishin mjaftueshëm të pastra.
- Në vitet 1940, teknologjia bazohej në **tubat vakum** për përforcim dhe komutacion — këto ishin të mëdha, konsumonin shumë energji dhe ishin të paqëndrueshme. Kërkohej **një zëvendësues me përmasa të vogla, i besueshëm dhe energjitikisht efikas**.
- **1947:** Shpikja e parë praktike: BJT (Bardeen, Brattain, Shockley në Bell Labs).
- **1959:** Pasi u zgjidhën problemet me oksidet (SiO₂), u bë i mundur ndërtimi i MOSFET.
- Sot: MOSFET dominon elektronikën digjitale dhe IC-të (integrated circuits), BJT përdoret kryesisht në përforcues ose qarkë analogë me rrymë të madhe.

<img src="../figs/lec8/Triode_symbol.svg" alt="Triode" width="44%" style="background-color:#fff;" > <img src="../figs/lec8/Triode.svg" alt="Triode" width="45%" style="background-color:#fff;" >


**Ironi historike**:  
Pavarësisht se transistori BJT (1947) u zbulu përpara FET (1959), në fakt, ideja e parë ishte për një **Field Effect Transistor (FET)** (1925) — por atë kohë, nuk ishte e mundur për shkak të papërsosmërisë së materialeve.


<br>

## Kategoritë e Transistorëve

Dy janë tipet kryesore të transistorëve: 
1. Transistorë **Bipolare** (Bipolar Junction Transistor, BJT shkurt)
2. Transistorët me Efekt Fushe (Field Effect Transistor, FET, shkurt) që ndahen në
   - **JFET** (Junction Field Effect Transistor)
   - **MOSFET** (Metal Oxide Semiconductor Field Effect Transistor)


| Aspekti        | BJT                              | FET                         |
|----------------|-----------------------------------|------------------------------|
| Kontroll       | Rrymë ($I_B$)                    | Tension ($V_{GS}$)           |
| Bartës         | Elektrone dhe vrima (bipolar)     | Elektrone ose vrima (unipolar) |
| Mekanizmi      | Injeksion dhe difuzion             | Kontroll me fushë elektrike  |
| Impedanca Hyrëse | E ulët                           | Shumë e lartë                |
| Konsumi i energjisë | Më i lartë                   | Shumë i ulët                 |
| Materiali kryesor |  Si, Ge                        | Si, SiO₂                     |
| Përdorimi kryesor | Amplifikim dhe komutacion      | Elektronikë digjitale dhe përforcim me fuqi të ulët |


### Tre terminalet e një transistori:

- **Hyrje e energjisë** - Collector (C) në BJT, Drain (D) në FET
- **Dalje e energjisë** - Emitter (E) në BJT, Source (S) në FET

# **II.A — Struktura dhe Funksionimi Fizik i Transistorit Bipolar (BJT)**

## **1. Struktura Bazë e BJT**

<img src="../figs/lec8/npn_Cross-section.svg" alt="npn cross section" width="30%" style="background-color:#fff;" > <img src="../figs/lec8/BJT_pnp_npn.svg" alt="bjt" width="50%" style="background-color:#fff;" >

Një **transistor bipolar me bashkim** (BJT) përbëhet nga **tre rajone gjysmëpërçuese**:
* **Emiter (E)** – shumë i dopuar për të siguruar shumë bartës.
* **Bazë (B)** – e hollë dhe lehtë e dopuar, e cila lejon vetëm një pjesë të vogël të rekombinimit.
* **Kolektor (C)** – më pak i dopuar dhe më i gjerë, për të thithur bartësit.

BJT-të ndahen në dy lloje:
* **NPN**: n–p–n
* **PNP**: p–n–p

Në vijim do të fokusohemi tek tipi **NPN**, pasi është më i përdorur dhe më intuitiv për analizë. Analiza për transistorin PNP është identike, vetëm ndryshojmë polaritetin e tensioneve dhe kahun e rrymës. Çdo gjë tjetër në analizë është identike.


## **2. Injektimi dhe Transporti i Bartësve**

Në një transistor **NPN** që operon në **regjimin aktiv përpara**, kemi:
* Bashkimi **emiter-bazë** është i **polarizuar drejt**.
* Bashkimi **bazë-kolektor** është i **polarizuar në kundërt**.

Kjo gjeneron këtë seri ngjarjesh:
1. Elektronet injektohen nga **emiteri** (n) në **bazën** (p).
2. Baza është **shumë e hollë**, kështu që **shumica e elektroneve nuk rekombinohen**, por kalojnë përmes bazës.
3. Në anën tjetër, **fusha elektrike e bashkimit bazë-kolektor** i tërheq këto elektrone drejt **kolektorit**.

<img src="../figs/lec8/npn_Operation.svg" alt="npn cross section" width="80%" style="background-color:#fff;" >

<img src="../figs/lec8/NPN_Band_Diagram_Active.svg" alt="npn cross section" width="50%" style="background-color:#fff;" >


## **3. Rrymat përmes të tre terminaleve**

Në këtë proces, lindin tre rryma kryesore:
* **$I_E$**: rryma e emiterit — rrjedh jashtë nga emiteri.
* **$I_B$**: rryma e bazës — shumë e vogël (rreth 1% e $I_E$).
* **$I_C$**: rryma e kolektorit — gati e barabartë me $I_E$.

Për rrymat vlen:
$$ I_E = I_B + I_C $$

Rryma e kolektorit është:
$$ I_C = \beta I_B $$
ku $\beta$ është **fitimi i rrymës** (current gain), zakonisht $50 \leq \beta \leq 300$.


## **4. Thelbi fizik i përforcimit**

Amplifikimi ndodh sepse:
* Një **rrymë shumë e vogël** në bazë ($I_B$)
* **Kontrollon rrjedhën e një rryme shumë më të madhe** në kolektor ($I_C$)

Ky raport krijon efektin e **përforcimit të rrymës**.


<div style="border:5px solid #0a0; background-color:#ca0; color:#003; border-radius: 15px; padding: 10px;">

#### **Përmbledhje Strukturës & Funksionit**

* BJT përdor **injektimin e bartësve** dhe **fushën elektrike** për të kontrolluar rrjedhjen e rrymës.
* **Struktura asimetrike** (baza e hollë, emiteri i dopuar) është çelësi për amplifikim efikas.
* Ky funksion fizik bën të mundur modelimin si përforcues apo komutator.

</div>

<div style="border:3px solid #06c; background-color:#eef6ff; color:#002; border-radius: 12px; padding: 12px;">

#### **Mekanizmi fizik kryesor i BJT**

Në regjimin aktiv përpara:

1. Bashkimi **emiter–bazë** polarizohet drejt.
2. Emiteri injekton shumë bartës në bazë.
3. Baza është shumë e hollë dhe lehtë e dopuar, prandaj vetëm pak bartës rekombinohen.
4. Shumica e bartësve arrijnë te bashkimi bazë–kolektor.
5. Fusha elektrike në bashkimin bazë–kolektor i tërheq bartësit drejt kolektorit.

Prandaj një ndryshim i vogël në $V_{BE}$ ose $I_B$ shkakton një ndryshim shumë më të madh në $I_C$.

</div>


# **II.B — Regjimet e Funksionimit të Transistorit Bipolar (BJT)**

Transistori bipolar (BJT) ka **katër regjime kryesore funksionimi**, të përcaktuara nga **polarizimi i dy bashkimeve p-n**:
* **Bashkimi emiter–bazë (BE)**
* **Bashkimi bazë–kolektor (BC)**

Këto regjime janë të rëndësishme për të kuptuar **si përdoret BJT si përforcues apo si çelës**.

<img src="../figs/lec8/bjt_char.png" alt="npn cross section" width="50%" style="background-color:#fff;" >


## **Tabela përmbledhëse e regjimeve**


| Regjimi | Bashkimi emiter–bazë (BE) | Bashkimi bazë–kolektor (BC) | Përdorimi kryesor |
|---|---|---|---|
| **Cutoff / prerje** | polarizim i kundërt | polarizim i kundërt | çelës i hapur, $I_C \approx 0$ |
| **Aktiv përpara** | polarizim drejt | polarizim i kundërt | amplifikim linear |
| **Saturim** | polarizim drejt | polarizim drejt | çelës i mbyllur |
| **Aktiv i kundërt** | polarizim i kundërt | polarizim drejt | rrallë përdoret |



## **1. Regjimi Aktiv Përpara (Forward Active Region)**

* **BE** është i polarizuar **drejt**:
  $$ V_{BE} \gtrsim 0.6\ \text{V (për silicin)} $$
* **BC** është i polarizuar **në kundërt**:
  $$ V_{CB} > 0 $$

**Rezultati:**
* Elektronet injektohen nga emiteri në bazë.
* Kolektori i tërheq → rrjedh $I_C \approx \beta I_B$.
* Ky është regjimi **i dëshiruar për amplifikim**.

## **2. Regjimi i Saturimit**

* Të dy bashkimet **BE dhe BC** janë **të polarizuara drejt**.
  $$ V_{BE} \gtrsim 0.6\, \text{V},\quad V_{CB} < 0 $$

**Rezultati:**
* Zona e bazës është “e mbushur me bartës”.
* Transistori sillet si një **çelës i mbyllur**.
* Rryma $I_C$ është e kufizuar nga qarku → nuk ndjek më $I_C = \beta I_B$.


## **3. Regjimi i Prerjes (Cutoff Region)**

* Të dy bashkimet janë **të polarizuara në kundërt**:
  $$ V_{BE} < 0,\quad V_{CB} < 0 $$

**Rezultati:**
* Asnjë bartës nuk injektohet.
* Rryma $I_C \approx 0$ → **transistori është i fikur**.
* Përdoret si çelës **OFF** në logjikën digjitale.

## **4. Regjimi Aktiv i Kundërt**

* **BE** është në **kundërt** (reverse bias) dhe **BC** në **drejt** (forward bias).
* Struktura është e anasjelltë → kolektori vepron si emiter.

**Rezultati:**
* Funksionon, por **me efikasitet shumë më të ulët**:
  * $\beta_{\text{kthim}} \ll \beta_{\text{përpara}}$
* Përdoret rrallë → vetëm në analiza teorike ose tranzicion gjatë komutimit.

<div style="border:5px solid #0a0; background-color:#ca0; color:#003; border-radius: 15px; padding: 10px;">

#### **Përmbledhje e Regjimeve**

* Regjimi **aktiv përpara** është **zemra e përforcimit analog**.
* Regjimet **saturim** dhe **cutoff** janë themelore në **logjikën digjitale** (ON/OFF).
* Kuptimi i polarizimeve për secilën gjendje është thelbësor për projektim qarku.

</div>

#### **Kujdes terminologjik: saturimi te BJT**

Te **BJT**, saturimi do të thotë se të dy bashkimet p-n janë të polarizuara drejt. Në këtë regjim transistori nuk sillet më si përforcues linear dhe lidhja $I_C = \beta I_B$ nuk është më e vlefshme.

Në praktikë, saturimi i BJT përdoret zakonisht për ta përdorur transistorin si **çelës të mbyllur**.

</div>




<div style="border:3px solid #777; background-color:#f7f7f7; color:#111; border-radius: 12px; padding: 12px;">

#### **Hierarkia praktike e modeleve të BJT**

1. **Modeli DC i thjeshtuar**  
   Përdoret për pikën e punës: $V_{BE} \approx 0.7\,\text{V}$ dhe $I_C \approx \beta I_B$.

2. **Modeli eksponencial**  
   Përdoret për të kuptuar varësinë fizike në regjimin aktiv:
   $$
   I_C = I_S e^{V_{BE}/V_T}
   $$

3. **Modeli Ebers–Moll**  
   Model më i plotë për të gjitha regjimet e BJT.

4. **Modeli hybrid-$\pi$**  
   Përdoret për analizë të sinjalit të vogël pasi të jetë gjetur pika e punës.

</div>


# **II.C — Modelimi Matematikor i BJT: Modeli Ebers–Moll dhe Përafrimet**

## **Qëllimi i modelimit**

Modeli matematikor i transistorit bipolar shërben për të përshkruar **rrymat nëpër terminalet e tij** ($I_E$, $I_B$, $I_C$) në funksion të **tensioneve të aplikuara** dhe të **vetive fizike** të pajisjes.

Modeli **Ebers–Moll** është një nga më të rëndësishmit, pasi:

* Përshkruan të **gjitha regjimet** e funksionimit të BJT-së.
* Është bazuar në **transportin e bartësve** dhe **sjelljen e bashkimeve p-n**.
* Përdor konceptin e dy **rrymave të ngopjes** dhe dy **faktorëve të fitimit** (gain) 


## **1. Ekuacionet e plota të modelit Ebers–Moll**

<img src="../figs/lec8/Ebers-Moll_model_schematic_(NPN).svg" alt="npn cross section" width="40%" style="background-color:#fff;" > <img src="../figs/lec8/Ebers-Moll_model_schematic_(PNP).svg" alt="npn cross section" width="40%" style="background-color:#fff;" >

Për një transistor **NPN**, rrymat përshkruhen si më poshtë:

$$
\begin{aligned}
I_E &= I_{ES} \left( e^{\frac{V_{BE}}{V_T}} - 1 \right) - \alpha_R I_{CS} \left( e^{\frac{V_{BC}}{V_T}} - 1 \right) \\
I_C &= \alpha_F I_{ES} \left( e^{\frac{V_{BE}}{V_T}} - 1 \right) - I_{CS} \left( e^{\frac{V_{BC}}{V_T}} - 1 \right) \\
I_B &= I_E - I_C
\end{aligned}
$$

**Simbolet dhe kuptimet fizike**

| Simbol        | Kuptimi                                                           |
| ------------- | ----------------------------------------------------------------- |
| $I_{ES}$   | Rryma e ngopjes së bashkimit emiter–bazë                          |
| $I_{CS}$   | Rryma e ngopjes së bashkimit bazë–kolektor                        |
| $V_{BE}$   | Tensioni midis bazës dhe emiterit                                 |
| $V_{BC}$   | Tensioni midis bazës dhe kolektorit                               |
| $V_T$      | Tensioni termik, $V_T = \frac{kT}{q} \approx 25.85\, \text{mV}$ |
| $\alpha_F$ | Fitimi përpara (forward gain) i bartësve (zakonisht $0.98 - 0.998$)            |
| $\alpha_R$ | Fitimi kthim (reverse gain) (zakonisht shumë më i vogël: $0.2 - 0.5$)          |


## **2. Përafrime në regjimin aktiv përpara**

<img src="../figs/lec8/Approximated_Ebers_Moll.svg" alt="npn cross section" width="40%" style="background-color:#fff;" >

Kur BJT është në **regjimin aktiv përpara**:
* $V_{BE} \gg V_T$ → $e^{\frac{V_{BE}}{V_T}} \gg 1$
* $V_{BC}$ është negativ (polarizim i kundërt) → $e^{\frac{V_{BC}}{V_T}} \approx 0$

### Thjeshtimet bëhen:
$$
\begin{aligned}
I_C &\approx \alpha_F I_{ES} e^{\frac{V_{BE}}{V_T}} \\
I_E &\approx I_{ES} e^{\frac{V_{BE}}{V_T}} \\
I_B &\approx I_E - I_C = (1 - \alpha_F) I_{ES} e^{\frac{V_{BE}}{V_T}}
\end{aligned}
$$


### Marrëdhënia me parametrin $\beta$

Shpesh preferohet të përdorim:
$$
\beta = \frac{\alpha_F}{1 - \alpha_F}
\quad \text{dhe} \quad
\alpha_F = \frac{\beta}{1 + \beta}
$$

Duke përdorur këto, kemi:
$$
I_C = \beta I_B \quad \text{dhe} \quad I_E = (1 + \beta) I_B
$$


## **3. Efektit Early dhe modeli i përmirësuar**

Kur rritet $V_{CE}$ në regjimin aktiv, zona e bazës zvogëlohet → rritet $I_C$ pak më shumë se sa do duhej sipas modelit eksponencial. Nga formula për $I_C = I_S\,\exp\left(\frac{V_{BE}}{V_T}\right)$ shohim që për një vlerë fikse $V_{BE}=\text{const}$ varësia e $I_C$ nga $V_{CE}$ është konstante. Kur $V_{CE}$ rritet shumë, baza efektivisht hollohet prej efektit Early dhe rrjedhimisht rritet rryma siç tregohet në figurën mëposhtë.

<img src="../figs/lec8/Early_effect_(NPN).svg" alt="npn cross section" width="40%" style="background-color:#fff;" >


### Korrigjohet me:
$$
I_C = I_S e^{\frac{V_{BE}}{V_T}} \left(1 + \frac{V_{CE}}{V_A}\right)
$$
ku:
* $V_A$ është **tensioni Early** (zakonisht $50 - 150$ V)
* Ky efekt është i rëndësishëm në **modelet e sinjalit të vogël**, duke përcaktuar rezistencën e daljes:
$$
r_o \approx \frac{V_A}{I_C}
$$


<img src="../figs/lec8/Early_effect_ICvsVCE).svg" alt="npn cross section" width="80%" style="background-color:#fff;" >



<div style="border:5px solid #0a0; background-color:#ca0; color:#003; border-radius: 15px; padding: 10px;">

####  **Përmbledhje e Modelit Ebers–Moll**

* Modeli **Ebers–Moll** përshkruan plotësisht sjelljen e një BJT-je në të gjitha regjimet.
* Për analiza të zakonshme (regjimi aktiv përpara), përdorim thjeshtuara që fokusohen vetëm te bashkimi emiter–bazë.
* **Efektit Early** përshkruan ndryshimin e gjërësisë efektive të bazës së BJT me ndryshimin e tensionit midis bazës dhe kolektorit $V_{CB}$ (kur $V_{BE}=\text{const}$).
* Efekti **Early** është përmirësim që saktëson përllogaritjet e rrymave te transistorit në qark real.

</div>

# **II.D — Modelimi i Sinjalit të Vogël për BJT**


Modelimi i sinjalit të vogël është thelbësor për të kuptuar dhe analizuar se si BJT vepron **si përforcues**, kur sinjalet që i aplikohen janë **shumë më të vogla se tensionet DC të polarizimit**.


## **Qëllimi**

* **Përshkrim linear** i transistorit rrotull pikës së punës (dmth Q-point).
* Llogaritja e: **fitimit të tensionit** (gain), **impedancës së hyrjes/daljes**, dhe **reagimit në frekuencë**.
* Modelet bazohen në **diferencimin e modeleve jo-lineare**.


<div style="border:5px solid #0a0; background-color:#ddd; color:#003; border-radius: 15px; padding: 10px;">


<img src="../figs/lec8/H_pi_model.svg" alt="npn cross section" width="50%" style="background-color:#fff;" >

### **Derivimi i ekuacioneve** me sinjal të vogël në regjimin aktiv përpara

Supozoni një amplifikator me BJT me emiter të përbashkët që është vendosur në një Q-point (quiscent point) ku $V_{BE}=V_{BE}^{\text{DC}}$ dhe $I_C=I_C^{\text{DC}}$. Tani supozojmë që tensioni i bazës perturobohet me një tension të vogël $v_{be}(t)\ll V_{BE}^{\text{DC}}$ (**perturbim i vogël sinusoidal**) i cili shkakton një ndryshim të rrymës së kolektorit me nje vlere $i_c(t) \ll I_C^{\text{DC}}$. Rrjedhimisht kemi:
$$
V_{BE}(t) = V_{BE}^{\text{DC}} + v_{be}(t)
$$

$$
I_C(t) = I_C^{\text{DC}} + i_c(t)
$$

Nga lidhja që kemi përmendur më lartë:
$$
I_C = I_S e^{\frac{V_{BE}}{V_T}}
$$

Diferencojmë $I_C$ në lidhje me $V_{BE}$ për të marrë përbërësin AC (sinjal të vogël):
$$
\frac{dI_C}{dV_{BE}} = \frac{d}{dV_{BE}} \left(I_S e^{\frac{V_{BE}}{V_T}}\right) = \frac{I_S}{V_T} e^{\frac{V_{BE}}{V_T}} = \frac{I_C}{V_T}
$$

Pra:
$$
i_c(t) = \frac{dI_C}{dV_{BE}} \cdot v_{be}(t) = \frac{I_C}{V_T} \cdot v_{be}(t)
$$

**Përkufizojmë transkonduktancën $g_m$**:
$$
g_m = \frac{dI_C}{dV_{BE}} = \frac{I_C}{V_T}
$$
Prandaj:
$$
i_c = g_m\; v_{be}
$$

#### **Impedanca e hyrjes baza–emiter: $r_\pi$**

$r_\pi$ është **rezistencë diferenciale për sinjal të vogël**, jo raport DC $V_{BE}/I_B$. Nga përkufizimi:
$$
i_b = \frac{i_c}{\beta} = \frac{g_m v_{be}}{\beta}
\Rightarrow v_{be} = i_b \cdot \left(\frac{\beta}{g_m}\right)
$$

Prandaj:
$$
r_\pi = \frac{\partial v_{be}}{\partial i_b}
      = \frac{\beta}{g_m}
      = \frac{\beta V_T}{I_C}
      = \frac{V_T}{I_B}
$$


#### **Modifikimet prej Efektit Early:**

$$
I_C = I_S e^{\frac{V_{BE}}{V_T}} \left(1 + \frac{V_{CE}}{V_A} \right)
$$


**Linearizimi rreth pikës Q**

Le të shkruajmë $V_{CE}$ si:

$$
V_{CE}(t) = V_{CE}^{\text{DC}} + v_{ce}(t)
$$

Duke zbërthyer $I_C$ në seri Taylor-i rreth pikës Q:

$$
i_c(t) = \frac{\partial I_C}{\partial V_{BE}} v_{be}(t) + \frac{\partial I_C}{\partial V_{CE}} v_{ce}(t)
$$

Derivimi:
1. Derivati ndaj $V_{BE}$ (si më parë):
$$
\frac{\partial I_C}{\partial V_{BE}} = \frac{I_C}{V_T} = g_m
$$
2. Derivati ndaj $V_{CE}$:
$$
\frac{\partial I_C}{\partial V_{CE}} = \frac{I_S}{V_A} e^{\frac{V_{BE}}{V_T}} = \frac{I_C}{V_A}
$$

$$
i_c = g_m v_{be} + \frac{1}{r_o} v_{ce}
\quad \text{ku} \quad r_o \approx \frac{V_A}{I_C}
$$

## Amplifikatori me emiter të përbashkët

Amplifikimi në tension përkufizohet si
$$ A_v = \frac{V_{\rm out}}{V_{\rm in}}, $$

ndërsa amplifikimi në rrymë përkufizohet si
$$ A_i = \frac{I_{\rm out}}{I_{\rm in}}. $$

Le ta aplikojmë këto në rastin e një BJT npn me emiter të përbashkët (dmth emiteri lidhet me sinjalin hyrës dhe atë dalës), si në figurë

<img src="../figs/lec8/NPN_common_emitter.svg" alt="npn cross section" width="20%" style="background-color:#fff;" > <img src="../figs/lec8/NPN_common_emitter_degeneration.svg" alt="npn cross section" width="20%" style="background-color:#fff;" >

Në të dy skemat kemi neglizhuar detajet e biasing (pra të vendosurit të transistorit në një pikë DC pune stabël, Q-point). Në rastin e dytë kemi shtuar një rezistencë në emiter dhe ky rast quhet me **emiter të degjeneruar**. **Ky konfigurim zvogëlon përforcimin por rrit linearitetin dhe stabilitetin e transistorit**.

$$ A_v=\frac{-g_m\, R_C}{g_m\,R_E + 1} $$

Në rastin e parë kur rezistenca e emiterit mungon kjo kthehet në $A_v=-g_m\,R_C$. Në rastin tjetër kur $g_m\,R_E \gg 1$, që arrihet për $R_E$ relativisht të mëdha, amplifikimi i tensionit ështv $A_v=-R_C/R_E$. Ndërsa amplifikimi i rrymës në secilin rast është:
$$ A_i=\beta$$

### Përmbledhje

* **$g_m = \dfrac{I_C}{V_T}$**
* **$r_\pi = \dfrac{\beta}{g_m} = \dfrac{\beta V_T}{I_C}$**
* **$r_o = \dfrac{V_A}{I_C}$**
* **$i_b = \dfrac{v_{be}}{r_\pi}$**
* **$i_c = g_m\, v_{be} + \dfrac{v_{ce}}{r_o}$**
* **$A_v=\frac{-g_m\, R_C}{g_m\,R_E + 1}$**
* **$A_i=\beta$**


<img src="../figs/lec8/Complete_common_emitter_amplifier.png" alt="npn cross section" width="90%" style="background-color:#fff;" >


</div>

## **2. Modeli me parametra h (h-parameter model)**

Modeli h përfaqëson transistorin si një **rrjet me dy porta**, duke lidhur tensionin dhe rrymën në hyrje me ato në dalje.

### Ekuacionet (konfigurimi me emiter të përbashkët):

<img src="../figs/lec8/BJT_h-parameters_(generalised).svg" alt="npn cross section" width="50%" style="background-color:#fff;" >

$$
\begin{aligned}
v_b &= h_{ie} i_b + h_{re} v_c \\
i_c &= h_{fe} i_b + h_{oe} v_c
\end{aligned}
$$

| Parametri   |                     Përkufizimi                   |
| ----------- | ----------------------------------------------- |
| $h_{ie}$ | Impedanca e hyrjes                                |
| $h_{fe}$ | Përforcimi i rrymës ($\approx \beta$)               |
| $h_{oe}$ | Përcjellshmëria e daljes ($\approx \frac{1}{r_o}$) |
| $h_{re}$ | Tensioni i kthimit (zakonisht shumë i vogël)      |

Ky model përdoret më rrallë për analizë analitike, por është i dobishëm për karakterizim eksperimental.

#### **Strategjia praktike për analizën e një qarku me BJT**

1. Hiq sinjalin AC dhe analizo vetëm qarkun DC.
2. Gjej $V_B$, $V_E$, $I_E$, $I_C$ dhe $V_{CE}$.
3. Kontrollo regjimin e punës: cutoff, aktiv apo saturim.
4. Nëse transistori është në regjimin aktiv, ndërto modelin e sinjalit të vogël.
5. Llogarit $g_m$, $r_\pi$ dhe, nëse duhet, $r_o$.
6. Gjej amplifikimin, rezistencën hyrëse dhe rezistencën dalëse.
7. Kontrollo nëse sinjali mund të dalë pa prerje ose shtrembërim.

# **II.E — Aplikime Praktike të Transistorit BJT në Qark**


## **Qëllimi i këtij seksioni**

Të tregojmë **si përdoret një BJT në praktikë**, si **përforcues i sinjalit** ashtu edhe si **çelës** në qark elektronik. Këto dy funksione përbëjnë bazën e përdorimit të BJT në elektronikë **analoge** dhe **digjitale**.


## **1. BJT si Përforcues**

<img src="../figs/lec8/common_EBC.jpeg" alt="npn cross section" width="50%" style="background-color:#fff;" >

Transistori përdoret si **përforcues i tensionit** në konfigurime të ndryshme:

### **1.1 Përforcuesi me emiter të përbashkët (Common Emitter Amplifier)**

* Konfigurimi më i zakonshëm.
* Jep **përforcimin më të madh të tensionit** dhe **përmbysje faze** (180°).

#### Skema e përgjithshme:

* Hyrja aplikohet në **bazë**.
* Dalja merret nga **kolektori**.
* **Emiteri** është tokë për sinjalin AC (direkt ose përmes një kondensatori by-pass).

#### Perforcimi i tensionit:
$$
A_v = \frac{v_{out}}{v_{in}} \approx -g_m (R_C \parallel r_o)
$$

#### Karakteristika:
* Fitim i lartë.
* Përmbysje faze.
* Impedancë hyrëse $r_\pi$.
* Impedancë daljeje $r_o$.


### **1.2 Përforcuesi me kolektor të përbashkët (Common Collector / Buffer)**

* Hyrja në bazë, dalja nga emiter.
* **Fitimi i tensionit afër 1** (pra, nuk përforcon tensionin por përforcon rrymën).
* Impedancë hyrëse e lartë, impedancë daljeje e ulët.

#### Përforcimi:
$$
A_v \approx \frac{r_E}{r_E + r_\pi / (\beta + 1)} \approx 1
$$

#### Përdorim:
* Si **buffer** për ndarjen/deciftimin e qarqeve me ngarkesa të ndryshme.


### **1.3 Përforcuesi me bazë të përbashkët (Common Base Amplifier)**

* Hyrja në **emiter**, dalja në **kolektor**, baza është tokë për AC.
* Përforcim i lartë tensioni, **pa përmbysje faze**.
* Impedancë hyrëse shumë e ulët.

#### Përdorim:
* Përdoret në qarqe me **frekuenca të larta** (RF-radio frequency).


## **2. BJT si Komutator (Çelës)**

Kur përdoret si **çelës**, BJT operon në **dy regjime**:

* **Cutoff** → transistori është **i fikur**.
* **Saturim** → transistori është **i ndezur**.

### Aplikim tipik:

#### Qark komandues i LED:

* **Baza** lidhet përmes një rezistence me një sinjal kontrolli.
* Kur sinjali është **i lartë**: BJT futet në **saturim** → LED ndizet.
* Kur sinjali është **i ulët**: BJT në **cutoff** → LED fiket.


#### Përllogaritje praktike:

1. Zgjidh **rezistencën e bazës** për të dhënë rrymën e nevojshme $I_B$:
   $$ R_B = \frac{V_{in} - V_{BE}}{I_B}, \quad \text{me} \quad I_B = \frac{I_C}{\beta_{forced}} $$
2. Ku $\beta_{forced}$ është zakonisht më i vogël se vlera nominale, për siguri.


## **Shembuj të zakonshëm praktikë**

| Aplikimi               | Roli i BJT          | Regjimi          |
| ---------------------- | ------------------- | ---------------- |
| Qark përforcuesi audio | Përforcues tensioni | Aktiv përpara    |
| LED driver             | Çelës ON/OFF        | Saturim / Cutoff |
| Qark logjik TTL        | Portë logjike       | Komutacion       |
| Sensor aktiv           | Amplifikon sinjalin | Aktiv përpara    |


<div style="border:5px solid #0a0; background-color:#ca0; color:#003; border-radius: 15px; padding: 10px;">

####  **Përmbledhje**

* BJT mund të përdoret për **përforcim** (aktiv përpara) dhe **komutacion** (cutoff/saturim).
* Konfigurimi me **emiter të përbashkët** është më i fuqishmi për amplifikim.
* **Fitimi** dhe **impedanca** varen nga konfigurimi dhe polarizimi.
* Në përdorim si çelës, dimensionimi i rrymës së bazës është kyç për kontroll të saktë.


</div>

# **II.F — Leximi praktik i parametrave të një BJT nga datasheet-i**

Në analiza fillestare përdorim modele të thjeshta, por në projektim real duhet kontrolluar datasheet-i i komponentit. Parametrat më të rëndësishëm janë:

| Parametri | Kuptimi praktik |
|---|---|
| $V_{CEO}$ | tensioni maksimal kolektor–emiter kur baza është e hapur |
| $I_C$ max | rryma maksimale e kolektorit |
| $P_D$ | fuqia maksimale që mund të disipojë komponenti |
| $h_{FE}$ ose $\beta$ | amplifikimi DC i rrymës; ka shpërndarje të madhe nga komponenti në komponent |
| $V_{CE(sat)}$ | tensioni kolektor–emiter kur BJT përdoret si çelës i mbyllur |
| $f_T$ | frekuenca tranzicionale, e rëndësishme për sinjale të shpejta |
| $R_{\theta JA}$ | rezistenca termike paketë–ambient |

<div style="border:3px solid #777; background-color:#f7f7f7; color:#111; border-radius: 12px; padding: 12px;">

#### **Kujdes praktik**

Mos projekto një qark duke marrë $\beta$ si vlerë fikse. Në datasheet zakonisht jepet një interval i gjerë për $h_{FE}$. Për këtë arsye, qarqet e mira të biasing-ut duhet të varen sa më pak nga vlera ekzakte e $\beta$.

# **Përmbledhje e Leksionit 10**

Në këtë leksion u ndërtua zinxhiri konceptual:

$$
\text{struktura fizike} \rightarrow \text{regjimi i punës} \rightarrow \text{modeli} \rightarrow \text{qarku}
$$

Pikat kryesore:

- BJT është transistor bipolar sepse përfshin elektrone dhe vrima.
- Në regjimin aktiv përpara, bashkimi BE është drejt ndërsa bashkimi BC është në të kundërt.
- Në amplifikim përdorim zakonisht regjimin aktiv përpara.
- Në komutim përdorim cutoff dhe saturim.
- Analiza praktike fillon me pikën DC të punës dhe pastaj kalon te modeli i sinjalit të vogël.
- Parametrat më të rëndësishëm të sinjalit të vogël janë $g_m$, $r_\pi$ dhe $r_o$.


# **Ushtrime**

### **1. Përforcuesi në konfigurim Common-Emitter me biasim me ndarës tensioni**

**Problemi.** Jepet një BJT NPN në konfigurim common-emitter. Parametrat janë:

* $V_{CC} = 12\,\text{V}$
* $R_1 = 100\,\text{k}\Omega$
* $R_2 = 18\,\text{k}\Omega$
* $R_C = 4.7\,\text{k}\Omega$
* $R_E = 1\,\text{k}\Omega$
* $\beta = 100$
* $V_{BE} \approx 0.7\,\text{V}$

Gjeni:

a) pikën e punës $I_C$ dhe $V_{CE}$;  
b) parametrat $g_m$ dhe $r_\pi$;  
c) një përafrim të amplifikimit të tensionit nëse emiteri është i bypass-uar për sinjal AC.

---

**Zgjidhja.**

Së pari gjejmë tensionin në bazë nga ndarësi i tensionit:

$$
V_B = V_{CC}\frac{R_2}{R_1+R_2}
= 12\frac{18}{100+18}
\approx 1.83\,\text{V}
$$

Tensioni në emiter është:

$$
V_E = V_B - V_{BE} \approx 1.83 - 0.7 = 1.13\,\text{V}
$$

Prandaj:

$$
I_E = \frac{V_E}{R_E} \approx \frac{1.13}{1\,\text{k}\Omega}
= 1.13\,\text{mA}
$$

Për $\beta$ të madh:

$$
I_C \approx I_E \approx 1.13\,\text{mA}
$$

Tensioni në kolektor:

$$
V_C = V_{CC} - I_C R_C
\approx 12 - (1.13\,\text{mA})(4.7\,\text{k}\Omega)
\approx 6.69\,\text{V}
$$

Prandaj:

$$
V_{CE} = V_C - V_E \approx 6.69 - 1.13 = 5.56\,\text{V}
$$

Ky është një Q-point i arsyeshëm për amplifikim, sepse transistori është larg cutoff dhe larg saturimit.

Parametrat e sinjalit të vogël janë:

$$
g_m = \frac{I_C}{V_T}
\approx \frac{1.13\,\text{mA}}{25.9\,\text{mV}}
\approx 43.6\,\text{mS}
$$

$$
r_\pi = \frac{\beta}{g_m}
\approx \frac{100}{43.6\,\text{mS}}
\approx 2.3\,\text{k}\Omega
$$

Nëse emiteri është i bypass-uar për sinjal AC dhe neglizhojmë $r_o$, atëherë:

$$
A_v \approx -g_m R_C
\approx -(43.6\,\text{mS})(4.7\,\text{k}\Omega)
\approx -205
$$

Në praktikë ky amplifikim zvogëlohet nga ngarkesa, rezistenca e burimit, $r_o$ dhe nga fakti që sinjali nuk duhet ta çojë transistorin në cutoff ose saturim.

### **2. BJT si çelës për LED**

**Problemi.** Një mikrokontroller me dalje $5\,\text{V}$ komandon bazën e një BJT NPN për të ndezur një LED. Supozojmë:

* $V_{CC} = 5\,\text{V}$
* $I_C \approx 10\,\text{mA}$
* $V_{BE} \approx 0.7\,\text{V}$
* në saturim duam të përdorim $\beta_{forced} \approx 10$

Gjeni një vlerë të përafërt për rezistencën e bazës $R_B$.

**Zgjidhja.**

Për saturim të sigurt nuk përdorim $\beta$ nominal. Përdorim një vlerë të detyruar më të vogël:

$$
I_B \approx \frac{I_C}{\beta_{forced}}
= \frac{10\,\text{mA}}{10}
= 1\,\text{mA}
$$

Prandaj:

$$
R_B \approx \frac{V_{in}-V_{BE}}{I_B}
= \frac{5-0.7}{1\,\text{mA}}
\approx 4.3\,\text{k}\Omega
$$

Një vlerë standarde praktike është:

$$
R_B \approx 4.7\,\text{k}\Omega
$$
