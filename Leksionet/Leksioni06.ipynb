# Leksioni 6 — Dinamika e Qarqeve (Domeni Kohor)

## 1. Motivimi

Deri tani kemi studiuar qarqe të përbëra kryesisht nga rezistenca dhe burime tensioni ose rryme.  
**Këto sisteme janë statike: dalja përgjigjet menjëherë ndaj hyrjes, pa evolucion në kohë.**

Kjo ndryshon kur futim elementë që ruajnë energji, si kondensatorët dhe bobinat.  
**Këta elementë krijojnë dinamikë: përgjigja e qarkut evoluon në kohë.**

Në këtë leksion do të studiojmë sisteme të rendit të parë dhe të dytë, pra qarqe me një ose dy elementë që ruajnë energji.  
Qëllimi ynë është të kuptojmë se si këto sisteme reagojnë ndaj eksitimeve dhe si varet sjellja e tyre nga parametrat fizikë.

Duam të kuptojmë se si qarqet elektrike evoluojnë në kohë pas aplikimit të një sinjali hyrës.

**Pyetja kryesore:**  
Duke dhënë një qark dhe një hyrje, si ndryshojnë tensioni dhe rryma në kohë?

**Ideja kryesore:**  
Dinamika lind nga elementët që ruajnë energji (kondensatorë dhe bobina).



## 2. Elementët Themelorë (R, C, L)

* **Rezistenca (R)**
  * Marrëdhënia: $V = RI$
  * Roli: **shpërndan energji (dissipim)**
  * **Nuk krijon dinamikë**

* **Kondensatori (C)**
  * Marrëdhënia: $I = C \frac{dV}{dt}$
  * Roli: **ruan energji elektrike**

* **Bobina (L)**
  * Marrëdhënia: $V = L \frac{dI}{dt}$
  * Roli: **ruan energji magnetike**


> **Vini re:**  
> **Vetëm C dhe L sjellin varësi në kohë** → ato krijojnë dinamikën e qarkut.


<br>
<br>


## 3. Sistemet e Rendit të Parë

<div class="alert alert-info">

> **Përkufizim:**  
> **Një sistem është i rendit të parë nëse përmban një element që ruan energji.**

Sistemet tipike janë RC ose RL, me një burim energjie (elektrik në RC ose magnetik në RL).

* **Perspektiva e energjisë**
    * Kondensatori: $E=\frac{1}{2} C\, V^2$
    * Bobina: $E=\frac{1}{2} L\, i^2$

    **Dinamika = energji që ruhet dhe lirohet me kalimin e kohës**

* <u>Të gjitha sistemet e rendit të parë ndjekin:</u>
    $$ \boxed{\frac{dx}{dt} + \frac{x}{\tau} = \text{input}} $$
    **Të gjitha sistemet e rendit të parë shfaqin sjellje eksponenciale**

* **Konstanta kohore $\tau$** 
    * Mat shpejtësinë e përgjigjes së sistemit
    * $\tau$ e vogël → sistem i shpejtë
    * $\tau$ e madhe → sistem i ngadaltë

</div>

<br>
<br>


### Qarku RC

<img src="../figs/lec5/RC.svg" align="center" width="300"/>

* **Ekuacioni diferencial:**
    Nga ligjet e Kirchhoff-it (KVL dhe KCL) kemi:
    $$ V_{\rm in} = v_R(t) + v_C(t), \quad v_R(t) = R i(t), \quad i(t)=C \frac{dv_C}{dt} $$
    ku $V_{\rm in}$ është tensioni i burimit. Pra:
    $$ \boxed{ R C \frac{dv_C}{dt} + v_C(t) = V_{\rm in}} $$

* <u>Zgjidhja: **Sjellje eksponenciale**.</u>  
  Në rastin e ngarkimit të kondensatorit, $v_C(0)=0$, burimi është konstant $V_{\rm in}=V_0$, dhe qarku mbyllet në $t=0$. Marrim:
    $$ \boxed{v_C(t) = V_0 \left(1 - e^{-t/RC} \right)} \quad \boxed{i(t) = \frac{V_0}{R} e^{-t/RC}} $$

* **Konstanta kohore:** $\tau = R C$
* Në $t = \tau$ → sistemi arrin rreth $(1-e^{-1})\approx 63\%$ të vlerës finale.

<br>
<br>

<u>**Çfarë duhet të vëzhgohet**</u>
* Rritja e $R$ ose $C$ → **përgjigje më e ngadaltë**
* **Përgjigje eksponenciale**
* **Nuk ka oshilime**

```python
import numpy as np
import matplotlib.pyplot as plt

# Parametrat
V0 = 5  # Volt
R = 1000  # Ohm
C = 1e-6  # Farad
tau = R*C
t = np.linspace(0, 0.01, 1000)

# Voltage and Current
v_C = V0*(1 - np.exp(-t/tau))
i = (V0/R)*np.exp(-t/tau)

# Plots
plt.figure(figsize=(10,5))
plt.plot(1e3*t, v_C,'-k',lw=2, label=r'Capacitor Voltage ($V_C$)')
plt.plot(1e3*t, 1e3*i,'-r',lw=2, label=r'Current ($i$)')
plt.xlabel('time [ms]')
plt.ylabel('Voltage [V], Current [mA]')
plt.title('Capacitor charging in RC circuit')
plt.grid(True)
plt.legend()
plt.show
```

<br>
<br>
<br>

### Qarku RL

<img src="../figs/lec5/RL.svg" align="center" width="300"/>

* E njëjta strukturë matematikore:
  $$ V_{\rm in} = v_R(t) + v_L(t), \quad v_R(t) = R i(t), \quad v_L(t)=L\frac{di}{dt} $$
  $$ \boxed{\frac{L}{R}\frac{di}{dt} + i(t) = \frac{V_{\rm in}}{R}} $$

* <u>Zgjidhja: **Sjellje eksponenciale**.</u>  
  Kur qarku mbyllet papritur, rryma ndryshon nga $i(0)=0$ në vlerën maksimale të lejuar $I_0=V_{\rm in}/R=V_0/R$. Pra:
    $$ \boxed{i(t) = I_0 \left(1 - e^{-t/\tau} \right)}$$

* **Konstanta kohore:** $\tau = L/R$

Qarku RL ka të njëjtën strukturë si qarku RC. Rryma rritet gradualisht për shkak të induktancës.


* **Çfarë duhet të vëzhgohet**
  * Rritja e $L$ → **përgjigje më e ngadaltë**
  * Rritja e $R$ → **zbehje (damping) më e shpejtë**
  * Sjellja mbetet **eksponenciale**


### Vëreni që:
**Qarqet RC dhe RL shfaqin të dyja përgjigje eksponenciale të përcaktuara nga një konstante e vetme kohore.**

<br>

```python
# Parameters
V0 = 5  # Volt
R = 1000  # Ohm
L = 0.5  # Henry
tau = L/R
t = np.linspace(0, 0.01, 1000)

# Current and Voltage
i_L = (V0/R)*(1 - np.exp(-t/tau))
v_L = V0 * np.exp(-t/tau)

# Plots
plt.figure(figsize=(10,5))
plt.plot(1e3*t, v_L,'-k',lw=2, label=r'Inductor Voltage ($V_L$)')
plt.plot(1e3*t, 1e3*i_L,'-r',lw=2, label=r'Inductor Current ($i$)')
plt.xlabel('Time [ms]')
plt.ylabel('Voltage [V], Current [mA]')
plt.title('Inductor Charging in RL Circuit')
plt.grid(True)
plt.legend()
plt.show()
```

## 5. Sistemet e Rendit të Dytë (RLC)

<div class="alert alert-info">

> **Përkufizim:**  
> **Një sistem është i rendit të dytë nëse përmban dy elementë që ruajnë energji.**

Sistemet tipike janë qarqet LC ose më të përgjithshme RLC. Shkëmbimi i energjisë midis dy burimeve mund të prodhojë oshilime.


* <u>Të gjitha sistemet e rendit të dytë ndjekin:</u>
    $$ \boxed{\frac{d^2x}{dt^2} + 2\gamma\frac{dx}{dt} + \omega_0^2 x = \text{hyrja}} $$
    ku:
    * $\omega_0$ është frekuenca natyrore e sistemit  
    * $\gamma$ është faktori i zbehjes (damping)


##### Regjimet e Sjelljes (Përgjigja e sistemit)

* **Nën-kritik (Underdamped)** ($\gamma < \omega_0$)
    * **Përgjigje oshiluese**
    * Energji shkëmbehet midis L dhe C

* **Kritikisht i frenuar (Critically damped)** ($\gamma = \omega_0$)
    * Rikthimi më i shpejtë në ekuilibër  
    * Pa oshilime

* **Mbi-kritik (Overdamped)** ($\gamma > \omega_0$)
    * Zbehje e ngadaltë  
    * Pa oshilime


##### Figura Fizike

* Kondensatori ↔ energji elektrike  
* Bobina ↔ energji magnetike  
* Rezistenca → shpërndan energji  

**Sjellja e sistemit = balancë midis shkëmbimit të energjisë dhe shpërndarjes së saj**

</div>

<br>
<br>



### Qarku RLC (në seri)

<img src="../figs/lec5/RLC_seri.svg" align="center" width=500/>

**Ekuacioni diferencial**

Nga ligji i Kirchhoff-it për tensionin (KVL):
$$ V_{\rm in}(t) = v_R(t) + v_L(t) + v_C(t) $$
$$ V_{\rm in}(t) = R i(t) + L\frac{di(t)}{dt} + \frac{1}{C}\int i(t)\, dt $$

Duke shkruar rrymën si $i(t) = \frac{dq(t)}{dt}$, ku $q(t)$ është ngarkesa, marrim:
$$ \boxed{L\frac{d^2q}{dt^2} + R \frac{dq}{dt} + \frac{1}{C} q = V_{\rm in}} $$

Forma standarde e rendit të dytë.



### Regjimet

* Nën-kritik → sjellje oshiluese  
* Kritik → përgjigje më e shpejtë pa oshilim  
* Mbi-kritik → frenim e ngadaltë  

```python
import numpy as np
import matplotlib.pyplot as plt
from ipywidgets import interact, FloatSlider

# function that computes current
def rlc_current(R, L, C):
    V0 = 10  # Amplitude of the source (Volt)
    C  *= 1e-6 # Convert microFarad in Farad
    f = np.logspace(1, 4, 1000)  # Frequency (Hz)
    omega = 2 * np.pi * f

    Z = R + 1j * (omega * L - 1/(omega * C))
    I = V0 / np.abs(Z)

    plt.figure(figsize=(9,5))
    plt.plot(f, I, '-k', lw=2)
    plt.title('Response in frequency of the RLC circuit')
    plt.xlabel('Frekuency [Hz]')
    plt.ylabel('Current [A]')
    plt.grid(True)
    plt.ylim(0, np.max(I)*1.1)
    plt.xscale('log')
    plt.show()

# Interactive Sliders for R, L and C
interact(rlc_current,
         R=FloatSlider(min=10, max=1000, step=10, value=100, description=r'$R (\Omega)$'),
         L=FloatSlider(min=0.01, max=1.0, step=0.01, value=0.1, description=r'$L (H)$'),
         C=FloatSlider(min=0.1, max=100, step=0.1, value=10, description=r'$C (\mu F)$')
        );
```


### Interpretimi fizik

* Shkëmbim energjie midis L dhe C  
* Rezistenca shpërndan energji  


## 6. Simulimi & Vëzhgimi

Për çdo simulim (RC, RL, RLC), identifikoni:
* Konstanta kohore ose frekuenca e oshilimit  
* Lloji i përgjigjes (eksponenciale vs oshiluese)  
* Efekti i parametrave  

<u>Vëzhgime tipike:</u>
* **Rritja e $R$ → më shumë zbehje (damping)**  
* **Rritja e $C$ ose $L$ → dinamikë më e ngadaltë**  
* **RLC → oshilon vetëm nëse zbehja është e vogël**  


## 7. Përmbledhje

* **Rezistenca → nuk krijon dinamikë (vetëm shpërndan energji)**  
* **Kondensatori & Bobina → krijojnë dinamikë**  
* **Sistemet e rendit të parë → përgjigje eksponenciale**  
* **Sistemet e rendit të dytë → oshilime + zbehje**  
* **Rendi i sistemit = numri i elementëve që ruajnë energji**  
* RC/RL → dinamikë eksponenciale  
* RLC → oshilime + zbehje

# Lexim Opsional: Memristori — Elementi i Katërt

<img src="../figs/lec5/Two-terminal_non-linear_circuit_elements.svg" align="center" width=500/>

Në teorinë klasike të qarqeve kemi tre elementë themelorë:
- Rezistenca (R): lidh tensionin me rrymën  
- Kondensatori (C): lidh tensionin me ngarkesën  
- Bobina (L): lidh rrymën me fluksin magnetik  

Ekziston edhe një element i katërt i parashikuar:

### Memristori (M)
- Lidh: **ngarkesën ($q$)** dhe **fluksin magnetik ($\\varphi$)**  
- Ideja: një <u>**rezistencë me memorie**</u>  


### Sjellja fizike (intuitive)
- Rezistenca varet nga historiku i rrymës  
- Kur rryma ndalon → gjendja **ruhet** (rezistenca mbetet e njëjtë)  
- Pra: një komponent **jo-fluturues (non-volatile)**  


### Pse është interesant?
- Parashikuar nga **Leon Chua (1971)**  
- Realizuar eksperimentalisht rreth **2008 (HP Labs)**  
- Zbatime të mundshme:
  - **memorie jo-fluturuese**  
  - **sisteme neuromorfike (imitojnë sinapset biologjike)**  
  - **hardware për inteligjencë artificiale (AI)**  


### Për këtë kurs
Ne do të fokusohemi vetëm në **R, L dhe C**.  
Memristori është një **shembull modern** që tregon se teoria e qarqeve vazhdon të evoluojë.


Ref: [HP 2008 Nature paper](https://web.ece.ucsb.edu/~strukov/papers/2008/Nature2008.pdf), [Wikipedia](https://en.wikipedia.org/wiki/Memristor)

## Lexim Opsional: Transformimi Laplace & Impedanca e Përgjithësuar

### Ideja kryesore

Gjatë analizës së qarqeve, shpesh marrim **ekuacione diferenciale** (për shkak të C dhe L).

Transformimi Laplace i shndërron ato në **ekuacione algjebrike**, të cilat janë shumë më të lehta për t’u zgjidhur.

---

### Rezultati kryesor

Në domenin Laplace (me variablin $s$), elementët e qarkut sillen si:

- Rezistenca: $Z_R = R$  
- Kondensatori: $Z_C = \\frac{1}{sC}$  
- Bobina: $Z_L = sL$  

Kjo na lejon t’i trajtojmë qarqet si **rrjete algjebrike të thjeshta**.

---

### Pse është e fuqishme

- Ekuacionet diferenciale → algjebër  
- Kushtet fillestare përfshihen natyrshëm  
- Funksionon për **çdo hyrje** (jo vetëm sinusoide)  

---

### Kuptimi fizik

Variabli i Laplace-it:
$$
s = \\sigma + j\\omega
$$

- $\\sigma$ → pjesa tranziente (rritje/zbehje eksponenciale)  
- $\\omega$ → oshilimi (frekuenca)  

Laplace përfshin **si sjelljen tranziente ashtu edhe atë stacionare**.

---

### Lidhja me analizën AC

Pas një kohe të gjatë ($t \\gg \\tau$), komponentët tranzientë zhduken:

$$
s \\rightarrow j\\omega
$$

Pra, Laplace reduktohet në **analizën Fourier (AC)**.

---

### Për këtë kurs

Ne **nuk do të përdorim plotësisht llogaritjet me Laplace**.

Do ta përdorim vetëm për të kuptuar:
- nga vjen impedanca  
- pse funksionon analiza AC
