# **1. Nga bashkimi p–n tek dioda**

Në leksionin e mëparshëm (Leksioni08), studiuam se si formohet një **bashkim p–n** dhe si sillen bartësit e ngarkesës në zonën e kontaktit.

Këtu bëjmë hapin tjetër:
> Ne e trajtojmë bashkimin p–n si një **element qarku**, të quajtur **diodë**.

<div style="border:5px solid #0a0; background-color:#ca0; color:#003; border-radius: 15px; padding: 10px;">

### **Ideja kryesore**

Një diodë është **komponenti elektronik jolinear më i thjeshtë**.

Ndryshe nga një rezistencë, e cila ndjek një marrëdhënie lineare $I \propto V$, një diodë shfaq një **karakteristikë shumë jolineare e rrymë–tension (V–I)**:
* Përcjell ndjeshëm vetëm në një drejtim (polarizimi i drejtë)
* Bllokon rrymën në drejtimin e kundërt (polarizimi i kundërt), deri në shkatarrim.

</div>

### **Nga fizika në pajisje**

Proceset mikroskopike të studiuara në Leksionin08 (difuzioni, drift-i, rajoni i zbrazjes (zona e varfëruar), barriera potenciale) çojnë në një sjellje makroskopike:
* Polarizim i drejtë → barriera zvogëlohet → rryma rrjedh
* Polarizim i kundërt → barriera rritet → rryma zvogelohet

Kjo sjellje është ajo që e përcakton **diodën si element qarku**.

---

### **Çfarë ndryshon në këtë leksion**

Në këtë leksion, ndryshojmë fokusin:

* Jo: *pse lëvizin bartësit*
* Por: *si sillet pajisja në një qark*

Ne do të:

* përshkruajmë **karakteristikën V–I**
* prezantojmë **modele matematikore dhe të thjeshtuara**
* përdorim këto modele për të analizuar **qarqe reale**

# **2. Karakteristika V–I e diodës**

Sjellja e një diode përshkruhet plotësisht nga **karakteristika e saj rrymë–tension (V–I ose volt-amper)**:

* Boshti horizontal → tensioni aplikuar mbi diodë $V_D$
* Boshti vertikal → rryma që rrjedh nëpër diodë $I_D$

Kurba mund të ndahet në tre rajone:

### **(a) Rajoni i polarizimit të drejtë ($V_D > 0$)**

* Rryma është **shumë e vogël** për tensione të ulëta
* Pas një tensioni të caktuar (~0.6–0.7 V për silicin), rryma rritet **shumë shpejt**

Ideja kryesore:
> Dioda kalon nga “pothuajse OFF” në “ON” në një interval të ngushtë tensioni

### **(b) Rajoni i polarizimit të kundërt ($V_D < 0$)**

* Rrjedh vetëm një **rrymë shumë e vogël**
* Dioda sillet pothuajse si një **qark i hapur**

Ideja kryesore:
> Rryma e kundërt është pothuajse konstante dhe shumë e vogël

### **(c) Rajoni i shkatarrimit (breakdown)**

* Në tensione të mëdha negative, rryma rritet papritur
* Kjo quhet **prishje ose shkatarrim**

Dy raste:
* E padëshiruar (diodat normale → dëmtohet në këtë zonë)
* E qëllimshme (diodat Zener → rregullim tensioni)

## **2.2 Çfarë e bën këtë të veçantë?**

Ndryshe nga rezistencat: $I = \frac{V}{R}$ një diodë **nuk** ka një raport konstant midis $V$ dhe $I$.

Në vend të kësaj:
> Dioda është një **pajisje jolineare**: ndryshime të vogla në tension mund të prodhojnë ndryshime të mëdha në rrymë.

Kjo është arsyeja pse:
* teknikat lineare të analizës **nuk aplikohen drejtpërdrejt**
* nevojiten **modele përafruese**

## **2.3 Rend madhësie**
Diodë tipike silicit:
* Përçim i drejtë:
  * $V_D \approx 0.7\ \text{V}$ → rryma mund të jetë nga mA në A
* Rajoni i kundërt:
  * rryma $\sim \mu$A ose nA

Kjo do të thotë:
$$
\text{Rryma në polarizim të drejtë} \gg \text{Rryma në polarizim të kundërt}
$$

Kjo asimetri mundëson:
* radrizimin (drejtimin e rrymës alternative)
* sjelljen si ndërprerës tensioni

## **2.4 Perspektiva dinamike**

Kurba nuk është vetëm statike — ajo tregon si sillet dioda në një qark:
* Rritje e vogël në $V_D$ (rajoni përpara) → rritje e madhe në $I_D$
* Prandaj:
  * dioda mund të sillet si një **ndërprerës**
  * dioda mund të ndikojë fort në funksionimin e qarkut

Deri tani, përshkruam sjelljen **grafikisht**.

Pyetja tjetër:

> A mund ta përshkruajmë këtë kurbë matematikisht?

Kjo çon në **ekuacionin e diodës së Shockley-t**.

# **3. Ekuacioni i diodës së Shockley-t**

Karakteristika V–I e një diode mund të përshkruhet nga **ekuacioni i Shockley-t**:

$$ I_D = I_S\; \left(e^{\frac{V_D}{n V_T}} - 1\right)$$

## **3.1 Kuptimi i parametrave**

* $I_D$ → rryma e diodës
* $V_D$ → tensioni në diodë
* $I_S$ → **rryma e ngopjes** (shumë e vogël, zakonisht nA–μA)
* $n$ → faktori i idealitetit (zakonisht $1 \le n \le 2$)
* $V_T$ → **tensioni termik**

$$
V_T = \frac{kT}{q} \approx 25,\text{mV në temperaturë dhome}
$$

## **3.2 Çfarë na tregon ky ekuacion?**

Ky ekuacion shpjegon formën e kurbës V–I:

### **Polarizim i drejtpërdrejtë ($V_D > 0$)**

* Termi eksponencial dominon:
  $$ I_D \approx I_S\; e^{\frac{V_D}{n V_T}} $$

Rezultati:

> Rryma rritet **në mënyrë eksponenciale** me tensionin

### **Polarizim i kundërt ($V_D < 0$)**

* Termi eksponencial bëhet shumë i vogël:
  $$ I_D \approx -I_S $$

Rezultati:
> Rrymë **shumë e vogël dhe pothuajse konstante** në drejtim të kundërt

## **3.3 Ideja kryesore**
> Dioda është eksponenciale, jo me prag të fiksuar.

Vlera e përdorur shpesh “0.7 V ndezje” **nuk është fizikë fundamental** — është një **përafrim inxhinierik**.

## **3.4 Pse nuk përdorim drejtpërdrejt Shockley në qarqe**

Edhe pse i saktë, ky ekuacion është:
* jolinear
* i vështirë për t’u zgjidhur analitikisht në qarqe

Prandaj:
> Në analizën e qarqeve, e zëvendësojmë me **modele të thjeshtuara**

## **3.5 Polarizimi i kundërt dhe shkatarrimi (breakdown)**

#### **Kufizimet e modelit të Shockley-t**

Ekuacioni është i idealizuar dhe nuk përfshin:
* rezistencën dinamile (e rëndësishme për rryma të mëdha)
* sjelljen në rajonin e shkatarrimit (breakdown region)
* varësinë nga temperatura (e përfshirë në mënyrë implicite, por shpesh neglizhohet)
* papërsosmëritë reale të pajisjes

#### **Polarizimi i kundërt**
Në rajonin e polarizimit të kundërt, ekuacioni i Shockley-t parashikon:
$$ I_D \approx -I_S $$

Kjo sugjeron një **rrymë të vogël konstante**.
Megjithatë, kjo është e vlefshme vetëm deri në një tension të caktuar.

## **3.5.1 Fenomeni i shka*tarrimit (breakdown)*

<img src="../figs/lec7/IV_diode.png" width="30%" style="background-color:#fff;" >

Kur tensioni i kundërt tejkalon një vlerë kritike:

$$ |V_D| \ge V_{BR} $$

Rryma rritet **shumë shpejt**. Kjo quhet **shkatarrim** (breakdown).

## **3.5.2 Dy mekanizma fizikë**

Ekzistojnë dy mekanizma kryesorë të përcjellshmërisë në polarizim të kundërt:

### **(a) Efekti Zener**
* Ndodh në **tensione të ulëta** (~2–6 V)
* Fusha elektrike e fortë shkakton **tunelim kuantik** të elektroneve

Ideja kryesore:
> Elektronet kalojnë bashkimin pa pasur nevojë për energji termike

### **(b) Efekti Ortek**
* Ndodh në **tensione më të larta** (> ~6–8 V)
* Bartësit fitojnë energji dhe **jonizojnë atomet**, duke krijuar më shumë bartës në një efekt ortek

Ideja kryesore:
> Reaksion zinxhir → shumëzim i shpejtë i bartësve të ngarkesës

## **3.5.3 Interpretimi praktik**

Nga pikëpamja e qarkut:
* Para shkatarrimit → dioda bllokon rrymën
* Pas shkatarrimit → dioda përçon shume mire në polarizim të kundërt

## **3.5.4 Dallim i rëndësishëm**
**Diodë normale**
* Breakdown është **i padëshiruar**
* Mund të dëmtojë pajisjen nëse rryma nuk kufizohet

**Diodë Zener**
* Breakdown është **i qëllimshëm dhe i kontrolluar**
* Përdoret për **rregullim tensioni**

## **3.5.5 Kufizimi i ekuacionit të Shockley-t**
> Ekuacioni i Shockley-t **nuk përshkruan sjelljen në rajonin e breakdown (pra qofte efektin ortek qofte ate zener)**.

Kjo është arsyeja pse:
* karakteristika V–I duhet kuptuar përtej ekuacionit
* modelet duhet të përfshijnë breakdown **kur është e nevojshme**

Në praktikë:
* breakdown nuk varet vetëm nga tensioni
* Por nga **fuqia e çliruar**:
  $$ P = V \cdot I $$

Pa kufizim rryme:
* dioda → mbinxehet → shkatarrohet (digjet)

## **3.5.7 Polarizimi i drejtë**

Tani sjellja e plotë e diodës është e qartë:
* Rajoni me polarizim të drejtë → përcjellshmëri eksponencial
* Rajoni me polarizim të kundërt → përcjellshmëri e ulët
* Rajoni i Breakdown → përcjellshmëri e madhe në polarizim të kundërt

## **3.6 Kalimi drejt modeleve inxhinierike**

Tani e thjeshtojmë këtë sjellje eksponenciale në përafrime të përdorshme:
* Modeli i diodës ideale
* Modeli me prag (tension konstant)

Këto na lejojnë të analizojmë qarqe **pa zgjidhur ekuacione jolineare me eksponenciale**.

# **4. Modelet inxhinierike të diodës**

Ekuacioni i Shockley-t përshkruan me saktësi diodën, por **nuk është praktik** për shumicën e analizave të qarqeve.

Prandaj:
> Ne e zëvendësojmë sjelljen eksponenciale me **modele të thjeshtuara** që kapin veçoritë thelbësore.

## **4.1 Pse nevojiten modelet**

Nga seksioni i mëparshëm:

* Rajoni me polarizim të drejtë → rritje eksponenciale e rrymës
* Rajoni me polarizim të kundërt → pothuajse zero rrymë

Kjo sugjeron një përafrim më të thjeshtë:

> Dioda sillet pothuajse si një **çelës i kontrolluar nga tensioni**.

## **4.2 Modeli i diodës ideale**

### **Përkufizimi**

* Polarizim i drejtë → dioda përçon në mënyrë perfekte
* Polarizim i kundërt → dioda bllokon plotësisht

$$ \text{Përpara: } V_D = 0,\quad I_D > 0$$
$$ \text{Kundërt: } I_D = 0,\quad V_D < 0 $$

* Sillet si një **çelës i mbyllur** në polarizim të drejtë
* Sillet si një **çelës i hapur** në polarizim të kundërt

# **5. Llojet e diodave**

Ndërsa të gjitha diodat bazohen në të njëjtin **parim të bashkimit p–n**, ndërtimi i tyre është **optimizuar** për aplikime të ndryshme.

Prandaj, dioda të ndryshme shfaqin:

* intervale të ndryshme tensioni
* shpejtësi të ndryshme tranzicioni
* sjellje të ndryshme në breakdown
* emetim drite (psh. në rasting e LED)

## **5.1 Llojet më të zakonshme të diodave**

| Lloji                          | Shembull tipik | Veçori kryesore                   | Përdorimi tipik                 |
| ------------------------------ | -------------- | --------------------------------- | ------------------------------- |
| **Diodë sinjali**              | 1N4148         | Ndërrim i shpejtë, rrymë e ulët   | Përpunim sinjali, qarqe logjike |
| **Diodë radrizuese**          | 1N4007         | Përballon rrymë/tension të lartë  | Furnizime (AC → DC)             |
| **Diodë Zener**                | BZX55, 1N4733  | Tension zener i kontrolluar   | Rregullim tensioni              |
| **LED (diodë që emeton dritë)** | Të ndryshme    | Lëshon dritë në polarizim të drejte | Indikatorë, ekrane, ndriçim etj.    |

## **5.2 Dallimet kryesore**

Shënim: Në këtë fazë, më e rëndësishme janë parimet e funksionimit dhe jo emrat specifik te diodave.

#### **(a) Tensioni $V_t$ (i pragut)**

* Silici → ~0.7 V
* LED → 1.8–3.3 V (varet nga materiali/ngjyra)

#### **(b) Rryma maksimale**

* Sinjali → mA
* Drejtëzues → në rangun e amperëve

#### **(c) Tensioni i breakdonw**

* Diodë standarde → **nuk duhet arritur**
* Diodë Zener → **projektuar për të punuar këtu**

#### **(d) Shpejtësia e ndërrimit**

* E rëndësishme në qarqe me frekuencë të lartë
* Dioda të shpejta (p.sh. 1N4148) vs radrizues të ngadaltë (p.sh. 1N4007)

## **5.3 Sqarim konceptual i rëndësishëm**

> Lloji i diodës nuk ndryshon në thelb sjelljen fizike të tyre.

Të gjitha diodat ndjekin:

* karakteristikë eksponenciale I–V
* të njëjtat qasje modelimi

Ajo që ndryshon është:

* vlerat e parametrave
* rajoni i funksionimit

## **5.4 Qasja praktike e përzgjedhjes**

Kur zgjedhim një diodë, duhet të pyesim:

* Çfarë rryme duhet të përballojë?
* Çfarë tensioni të kundërt duhet të përballojë?
* A ka rëndësi shpejtësia e ndërrimit?
* A kërkohet emetim drite?



## **5.5 Kalimi tek aplikimet**

Tani që kuptojmë:

* si sillet një diodë (karakteristika V–I)
* si modelohet (ideal, me prag)
* çfarë llojesh ekzistojnë

mund ta përdorim në qarqe, p.sh.:

# **6. Radrizimi: Konvertimi AC në DC**

Një nga përdorimet më të rëndësishme të diodës është **radrizimi**:

> Konvertimi i një sinjali alternativ (AC) në një sinjal të vazhduar (DC).


## **6.1 Radrizues i gjysmëvalës**

### **Qarku**

<img src="../figs/lec7/rectifier_halfWave.png" width="50%" style="background-color:#fff;" >

### **Funksionimi (modeli me prag)**

Supozoni:

* Diodë silici → $V_t \approx 0.7,\text{V}$
* Hyrja: $V_{in}(t) = V_0 \sin(\omega t)$

**Rasti 1: Gjysmëvalë pozitive** $ V_{in} > V_t$:

* Dioda është **ON**
* Dalja: $ V_{out} \approx V_{in} - V_t $

Sinjali kalon, por pak i reduktuar për shkak të $V_t$.

**Rasti 2: Gjysmëvalë negative** $ V_{in} < 0 $

* Dioda është **OFF**
* Dalja: $ V_{out} = 0 $

Sinjali bllokohet, nuk ka rrymë.


**Rezultati**

> Output-i përmban vetëm **gjysmëvalët pozitive** të hyrjes.


## **6.2 Ideja kryesore**

Dioda sillet si një **ndërprerës i varur nga polariteti**:

* ON → lejon kalimin e sinjalit
* OFF → e bllokon atë

Kjo është baza e:
* furnizimeve me energji
* formësimit të sinjalit

## **6.3 Interpretimi i formës së valës**

Duhet të mësoni të **lexoni qarqet në kohë**, jo vetëm përgjigjen statike në kohë:

* Hyrja: sinusoidale
* Dalja: **valë e prerë**

E rëndësishme:

* Dalja **nuk është DC konstante**
* Është **pulse DC**! (pulse pozitive)

## **6.4 Kufizimet e radrizuesve gjysmëvalë**

* Përdor vetëm gjysmën e sinjalit
* variacion i madh i sinjalit në dalje
* **Jo efikas!**

### **Rënia e tensionit**

* Rryma kalon nëpër **dy dioda**, prandaj:
  $$ V_{out} \approx V_{in} - 2,V_t $$

# **7. Qarqet kufizuese (Clipper)**

Një **clipper** është një qark që **kufizon (pret) tensionin** në një nivel të caktuar.

> Ai heq pjesët e sinjalit që kalojnë një prag të caktuar.

## **7.1 Ideja bazë**

* Kur dioda është **OFF** → sinjali kalon i pandryshuar
* Kur dioda është **ON** → tensioni **kufizohet**

Pra analiza është gjithmonë:

> Përcakto kur dioda është ON ose OFF

## **7.2 Clipper pozitiv**

### **Qarku**

<img src="../figs/lec7/clipper.svg" width="30%" style="background-color:#fff;" >

### **Funksionimi (modeli me prag)**

Supozoni: $V_t \approx 0.7,\text{V}$

**Rasti 1: $V_{in} < V_t$**

* Dioda është **OFF**
* Nuk rrjedh rrymë nëpër diodë
  $ V_{out} \approx V_{in} $
  Sinjali kalon i pandryshuar

**Rasti 2: $V_{in} \ge V_t$**

* Dioda është **ON**
* Dalja kufizohet:
  $V_{out} \approx V_t $

Sinjali **pritet në ~0.7 V**

**Rezultati**

> Dalja nuk mund të kalojë tensionin e pragut të diodës.

## **7.3 Clipper negativ**

**Koncepti**
Duke përmbysur diodën:

* Pjesa pozitive → e pandryshuar
* Pjesa negative → pritet

**Sjellja**

* Për hyrje negative më të vogla se $-V_t$ → dioda ON
* Dalja kufizohet pranë $-V_t$

## **7.4 Clipper me zhvendosje (prag i rregullueshëm)**

### **Ideja e qarkut**

Shto një burim DC $V_B$:

* Kufizimi ndodh në:
  $ V_{clip} \approx V_B + V_t $

## **7.5 Strategjia e përgjithshme e analizës**

Kur analizojmë qarqe me dioda duhet të ndjekim:

* **Hapi 1: Supozo gjendjen e diodës**
  * ON ose OFF
* **Hapi 2: Zëvendëso diodën me modelin**
  * ON → $V_D = V_t$
  * OFF → qark i hapur
* **Hapi 3: Zgjidh qarkun**
* **Hapi 4: Verifiko konsistencën**

A përputhet rezultati me supozimin?

Në fakt, kjo është një **metodë standarde për të gjitha qarqet jolineare**!

## **7.6 Ideja kryesore**

> Clipper-at nuk ndryshojnë të gjithë sinjalin — vetëm pjesët përtej pragut.

## **7.7 Aplikime praktike**

* Mbrojtje sinjali (kufizim tensioni)
* Formësim i valës
* Kondicionim analog i sinjalit

## **7.8 Kalimi tek seksioni tjetër**

Clipper-at **kufizojnë amplitudën**.

Më pas:

> Clamper-at **zhvendosin të gjithë nivelin e sinjalit**

# **8. Qarqet zhvendosëse (Clamper)**

Një **clamper** është një qark që **zhvendos të gjithë sinjalin lart ose poshtë** pa ndryshuar formën e tij.

> Ndryshe nga clipper-at, clamper-at nuk heqin pjesë të sinjalit — ato **shtojnë një offset DC**.

## **8.1 Ideja bazë**

Përdor:
* diodë
* kondensator
* rezistencë

Kondensatori ruan ngarkesë dhe **ripërcakton nivelin referencë** të sinjalit.

## **8.3 Funksionimi**

Supozoni hyrje sinusoidale.

* **Hapi 1: Gjysmëvalë negative**
  * Dioda bëhet **ON**
  * Kondensatori ngarkohet afërsisht në:  $V_C \approx V_{peak} - V_t$
    Kondensatori ruan tensionin

* **Hapi 2: Gjysmëvalë pozitive**
  * Dioda është **OFF**
  * Kondensatori sillet si një **burim tensioni në seri**
    $V_{out} = V_{in} + V_C$
    E gjithë vala zhvendoset **lart**

## **8.4 Rezultati**

> Sinjali zhvendoset në mënyrë që vlera minimale të jetë afërsisht $-V_t$ (ose pranë 0 sipas përafrimit).


## **8.5 Clamper negativ**

Duke përmbysur diodën:
* Kondensatori ngarkohet gjatë gjysmëvalës pozitive
* Sinjali zhvendoset **poshtë**

## **8.6 Kushti kryesor**

Për funksionim korrekt:
$$ \tau=RC \gg T $$

ku: $T$ = periudha e sinjalit hyrës

Kjo siguron:
* kondensatori **nuk shkarkohet ndjeshëm** midis cikleve

## **8.7 Ideja kryesore**

Mbani mend:
> Clipper-at modifikojnë amplitudën
> Clamper-at modifikojnë nivelin referencë

Mos i ngatërroni!

## **8.8 Aplikime praktike**

* Zhvendosje e nivelit DC
* Kondicionim sinjali para amplifikimit
* Qarqe televizive dhe komunikimi (historikisht të rëndësishme)

## **8.9 Lidhja me konceptet e mëparshme**

Funksionimi i clamper-it varet nga:

* ndërrimi i diodës (ON/OFF)
* ruajtja e ngarkesës në kondensator
* konstantet kohore

Kjo lidhet me:

* qarqet RC
* sjelljen tranzitore

# Shembuj zbatimi të Diodave

### Zbritës Tensioni (voltage droper)

Sic kemi thënë më lart, kur mbi diodë kalon rrymë krijohet një **"rënie tensioni"** (diferencë potenciale) mbi të.  
- Për një diodë silici p–n, kjo rënie është rreth **0.6-0.7 V**.  
- Për diodat gjermaniumi është rreth **0.2 V**.  
- Për diodat Schottky është rreth **0.4 V**.

(**Shënim: Vlera e sakta të tensionit ndryshojnë pak në varësi të tipit të diodës**.)

<img src="../figs/lec7/DC_V_drop.svg" height="100" style="background-color:#fff;" >

Nëse vendosim **shumë dioda në seri**, **rëniet e tensionit shtohen**:  
$$
V_{\text{total}} = \sum V_{\text{diodave}}
$$

Ky fenomen përdoret për të ndërtuar **zbritës tensioni** (“voltage droppers”) në qarqe ku kërkohet një **diferencë e vogël dhe fikse tensioni** midis dy pjesëve.

Ndryshe nga rezistencat (që gjithashtu mund të ulin tensionin), përdorimi i diodave:
- **Harxhon më pak energji në nxehtësi**,  
- **Siguron një tension më të qëndrueshëm**, më pak të varur nga ndryshimet në rrymë.

**Shënim:** Lidhja në seri e disa diodave në polaritet të drejtë mund të zëvëndësohet me **një diodë Zener** per të arritur të njëjtin efekt të rënies së tensionit.

## Mbrojtja nga Transienteve me Dioda Fly-Back

Kur rryma që kalon përmes një bobine ndryshon shpejt, fusha magnetike përkatëse gjeneron një **ndryshim të madh të tensionit** ("voltage spikes" apo transient), që mund të jenë shumë të dëmshme për komponentet apo qarqet që lidhen me bobinën.

- Këto transiente mund të kenë amplituda prej **qindra ose mijëra voltësh**.
- Fenomeni është **veçanërisht i zakonshëm në bobinat e releve**.

Për të mbrojtur qarkun:
- Përdoret një **diodë**, e quajtur **diodë fly-back**, e cila vendoset **paralelisht me bobinën e relesë**.
- Kjo diodë **krijon një qark të shkurtër** për transient të tensionit dhe e humb energjinë në mënyrë të sigurtë.

Përveç mbrojtjes së qarqeve elektronike, dioda **mbron edhe kontaktet mekanike** të relesë nga goditjet e forta që shkaktohen gjatë këtyre spikave.

### Diodat kufizuese (clipper diode)

Në qarkun e një **kufizuesi me përshtatje**, **niveli maksimal i daljes** kufizohet në një vlerë që përcaktohet nga **rezistenca e një potenciometri**.

**Ideja themelore:**
- **Anoda e diodës** lidhet me daljen e sinjalit,
- **Katoda e diodës** lidhet me një tension të përcaktuar përmes potenciometrit,
- Ky tension duhet të jetë rreth **0.6 V më i ulët** se niveli maksimal i dëshiruar i daljes,  
  që të kompensohet **rënie e tensionit përpara** për diodën (rreth 0.6 V për silic).

**Roli i potenciometrit:**
- Përshtat tensionin në anën negative të diodës, duke përcaktuar pikën ku fillon kufizimi.

**Kërkesë shtesë:**
- Tensioni +V që furnizon qarkun duhet të jetë **rreth 1 V më i lartë** se tensioni maksimal i sinjalit hyrës, për të garantuar që dioda të punojë siç duhet në rajonin e përcjelljes.

## Ushtrimi 1: Matja e Karakteristikës V-I të Diodës

**Komponentët dhe Pajisjet:**
- Diodë silici 1N4148
- Rezistencë 1 kΩ
- Furnizim DC i rregullueshëm (0—1.5 V)
- 2 multimetra ose osciloskop

**Skema e qarkut:**
```
Vdc ─── R ───|>|─── GND
```

**Procedura:**
1. Lidheni qarkun sipas skemës.
2. Rritni gradualisht tensionin nga 0 V deri në 1.5 V.
3. Matni rrymën (nga tensioni mbi rezistencë) dhe tensionin mbi diodën.
4. Vizato grafikisht $V_D$ kundrejt $I_D$.

**Pyetje:**
- Në çfarë vlere të tensionit fillon dioda të përçjellë?
- A sillet qarku sipas ligjit të Shockley?



## Ushtrimi 2: Diodë Clipper

**Komponentët dhe Pajisjet:**
- Diodë 1N4148
- Rezistencë 1 kΩ
- Gjenerator funksioni (sinus 10 Vpp, 1 kHz)
- Osciloskop 2-kanalësh

**Skema e qarkut:**
```
Vin ~ ──+---|>|── GND
         |
         +--- R --- Vout
```

**Procedura:**
1. Apliko sinjal sinusoidal nga gjeneratori funksioneve.
2. Mat sinjalin në hyrje dhe dalje me osciloskop.
3. Vëre ku dioda fillon të presë sinjalin.

**Eksploroni:**
- Ndrysho drejtimin e diodës.
- Shto burim DC në seri për të ndryshuar pikën e cliping.

**Pyetje:**
- Si sillet forma e sinjalit?
- Përse quhet klipper?



## Ushtrimi 3: Stabilizator me Zener

**Komponentët dhe Pajisjet:**
- Diodë Zener (psh 5.1 V)
- Rezistencë 470 Ω – 1 kΩ
- Furnizim DC 0—12 V
- Osciloskop ose voltmetër

**Skema e qarkut:**
```
Vin ── R ──+--- Vout
             |
           [Zener]
             |
            GND
```

**Procedura:**
1. Rrit Vin nga 0 deri në 12 V.
2. Mat Vout.
3. Vëre ndryshimin e Vout kur Vin > 5.1 V (tensionin që ka dioda zener që keni zgjedhur).

**Pyetje:**
- Përse Zener mban tension konstant?
- Si ndikon vlera e rezistencës?



## Ushtrimi 4: Rektifikimi (drejtimi) me një Diodë të Vetme

**Komponentët dhe Pajisjet:**
- Diodë 1N4007
- Rezistencë 1 kΩ
- Gjenerator funksioni
- Osciloskop

**Skema e qarkut:**
```
AC ~ ──|>|──+--- R --- GND
               |
              Vout
```

**Procedura:**
1. Apliko sinjal AC.
2. Vëre formën e sinjalit në dalje.
3. Analizo gjysmë-periodën aktive.

**Pyetje:**
- Përse gjysma negative mungon?

## Ushtrimi 5: Mbrojtja nga mbi-voltazhi me Zener

**Komponentët dhe Pajisjet:**
- Diodë Zener

**Skema e qarkut:**

**Procedura:**
