# **I — Pika e Qetësisë (Q-point) dhe Polarizimi**

## **1. Çfarë është Q-point?**

**Pika e qetësisë** është gjendja DC e transistorit kur nuk ka sinjal hyrës AC.

Për BJT, Q-point zakonisht jepet nga:

$$
Q_{BJT} = (I_C, V_{CE})
$$

Për MOSFET, Q-point zakonisht jepet nga:

$$
Q_{MOSFET} = (I_D, V_{DS})
$$

Q-point përcakton në cilin regjim punon transistori dhe sa hapësirë ka sinjali përpara se të shfaqet **clipping**, saturim i thellë ose cutoff.

## **2. Pse duhet polarizimi?**

Sinjali hyrës është zakonisht AC dhe ndryshon me kohën. Por transistori është pajisje jolineare.  
Prandaj i vendosim më parë një gjendje DC të qëndrueshme:

$$
\text{qarku real} = \text{polarizim DC} + \text{sinjal i vogël AC}
$$

Pa polarizim korrekt, transistori mund të jetë:

* në **cutoff** dhe të mos përçojë;
* në **saturim të thellë** si çelës;
* në një pikë pune të paqëndrueshme ndaj temperaturës ose tolerancave të komponentëve;
* në një zonë ku sinjali shtrembërohet.

## **3. Ku duhet vendosur Q-point?**

Për një **amplifikator BJT**, Q-point vendoset zakonisht në **regjimin aktiv përpara**.

Për një **amplifikator MOSFET**, Q-point vendoset zakonisht në **regjimin saturation të MOSFET-it**.

Për një **çelës**, Q-point nuk vendoset në mes të zonës lineare. Çelësi kalon midis:

$$
\text{OFF} \leftrightarrow \text{ON}
$$

domethënë midis cutoff dhe saturimit të thellë për BJT, ose midis cutoff dhe triode me $R_{DS(on)}$ të vogël për MOSFET.

<div style="border:3px solid #b55; background-color:#fff3f0; color:#300; border-radius: 12px; padding: 12px;">

## Kujdes terminologjik: “saturim” te BJT dhe MOSFET

Termi **saturim** nuk ka të njëjtin kuptim fizik te BJT dhe MOSFET.

* Te **BJT**, saturimi do të thotë se të dy bashkimet janë të polarizuara përpara; transistori përdoret zakonisht si **çelës i mbyllur**.
* Te **MOSFET**, saturimi është regjimi ku kanali ngushtohet pranë drain-it dhe rryma kontrollohet kryesisht nga $V_{GS}$; ky është shpesh regjimi i **amplifikimit**.

Prandaj, gjatë analizës duhet specifikuar gjithmonë pajisja dhe regjimi fizik.

</div>


# **II — Polarizimi i BJT**

## **1. Polarizimi me ndarës tensioni**

Kjo është teknika më e përdorur për amplifikatorë të thjeshtë me BJT.

Skema përmban:

* $R_1$ dhe $R_2$ që vendosin tensionin e bazës $V_B$;
* $R_E$ që stabilizon rrymën e emiterit përmes feedback negativ;
* $R_C$ që kthen ndryshimet e rrymës së kolektorit në tension daljeje.

Në afrim të parë:

$$
V_B \approx V_{CC}\frac{R_2}{R_1 + R_2}
$$

$$
V_E \approx V_B - V_{BE}
$$

$$
I_E \approx \frac{V_E}{R_E}
$$

dhe, për $\beta$ mjaft të madh:

$$
I_C \approx I_E
$$

Tensioni kolektor–emiter është:

$$
V_{CE} = V_C - V_E
$$

ku

$$
V_C = V_{CC} - I_C R_C
$$

prandaj:

$$
V_{CE} = V_{CC} - I_C R_C - I_E R_E
$$

Nëse $I_C \approx I_E$:

$$
V_{CE} \approx V_{CC} - I_C(R_C + R_E)
$$

## **2. Kontrolli i regjimit të punës**

Pasi llogaritet Q-point, duhet kontrolluar nëse BJT është në regjimin aktiv përpara.

Për NPN në amplifikator:

$$
V_{BE} \approx 0.6-0.7\,\text{V}
$$

dhe duhet të kemi:

$$
V_C > V_B > V_E
$$

ose në mënyrë praktike:

$$
V_{CE} > V_{CE(sat)}
$$

zakonisht me një rezervë të mjaftueshme, jo thjesht pak mbi $0.2\,\text{V}$.

## **3. Pse përdoret $R_E$?**

Rezistenca e emiterit krijon feedback negativ.

Nëse temperatura rritet, $I_C$ tenton të rritet. Por rritja e $I_E$ rrit edhe $V_E = I_E R_E$, duke ulur efektivisht $V_{BE}$. Kjo e kundërshton rritjen e rrymës.

Pra $R_E$ e bën Q-point më të qëndrueshme ndaj:

* ndryshimeve të temperaturës;
* variacioneve të $\beta$;
* ndryshimeve midis transistorëve realë.

<div style="border:5px solid #0a0; background-color:#ca0; color:#003; border-radius: 15px; padding: 10px;">

#### **Përmbledhje për BJT**

Për një amplifikator BJT, polarizimi i mirë nuk duhet të mbështetet fort te $\beta$, sepse $\beta$ ndryshon shumë nga një transistor në tjetrin.  
Teknika më e qëndrueshme është zakonisht **ndarësi i tensionit në bazë + rezistencë në emiter**.

</div>


## **4. Shembull: Projektim i Q-point për një BJT**

Të projektohet një pikë pune për një amplifikator me emiter të përbashkët.

Të dhënat:

* $V_{CC} = 12\,\text{V}$
* $R_1 = 100\,\text{k}\Omega$
* $R_2 = 18\,\text{k}\Omega$
* $R_C = 4.7\,\text{k}\Omega$
* $R_E = 1\,\text{k}\Omega$
* $V_{BE} \approx 0.7\,\text{V}$

### **Hapi 1 — Tensioni i bazës**

$$
V_B = 12\,\text{V}\cdot \frac{18}{100+18}
$$

$$
V_B \approx 1.83\,\text{V}
$$

### **Hapi 2 — Tensioni i emiterit**

$$
V_E = V_B - V_{BE}
$$

$$
V_E \approx 1.83 - 0.7 = 1.13\,\text{V}
$$

### **Hapi 3 — Rryma e emiterit dhe kolektorit**

$$
I_E = \frac{V_E}{R_E}
$$

$$
I_E \approx \frac{1.13\,\text{V}}{1\,\text{k}\Omega} = 1.13\,\text{mA}
$$

$$
I_C \approx I_E \approx 1.13\,\text{mA}
$$

### **Hapi 4 — Tensioni në kolektor**

$$
V_C = V_{CC} - I_C R_C
$$

$$
V_C \approx 12 - (1.13\,\text{mA})(4.7\,\text{k}\Omega)
$$

$$
V_C \approx 6.69\,\text{V}
$$

### **Hapi 5 — Tensioni kolektor–emiter**

$$
V_{CE} = V_C - V_E
$$

$$
V_{CE} \approx 6.69 - 1.13 = 5.56\,\text{V}
$$

Kjo është një Q-point e arsyeshme për amplifikim, sepse transistori është larg cutoff dhe larg saturimit të thellë.


# **III — Polarizimi i MOSFET**

## **1. Polarizimi me rezistencë në source**

Për një MOSFET n-channel, nëse porta vendoset në një tension $V_G$ dhe source ka rezistencë $R_S$, atëherë:

$$
V_S = I_D R_S
$$

dhe

$$
V_{GS} = V_G - V_S = V_G - I_D R_S
$$

Në regjimin saturation të MOSFET-it, modeli i thjeshtë jep:

$$
I_D = \frac{1}{2}k(V_{GS} - V_{th})^2
$$

Kombinimi i këtyre dy marrëdhënieve jep një ekuacion jolinear:

$$
I_D = \frac{1}{2}k(V_G - I_D R_S - V_{th})^2
$$

Prandaj për MOSFET shpesh Q-point gjendet me:

* zgjidhje numerike;
* grafikë të ngarkesës;
* SPICE;
* iterim me kalkulator ose Python.

## **2. Kontrolli i regjimit të punës**

Pasi gjendet $I_D$, llogariten:

$$
V_S = I_D R_S
$$

$$
V_D = V_{DD} - I_D R_D
$$

$$
V_{DS} = V_D - V_S
$$

Për një MOSFET n-channel në saturation duhet:

$$
V_{GS} > V_{th}
$$

dhe

$$
V_{DS} \geq V_{GS} - V_{th}
$$

Nëse kjo nuk plotësohet, MOSFET-i është në regjimin triode/ohmik, jo në saturation.

<div style="border:3px solid #b55; background-color:#fff3f0; color:#300; border-radius: 12px; padding: 12px;">

## Kujdes praktik për $V_{th}$

$V_{th}$ nuk është tensioni ku MOSFET-i është “plotësisht i ndezur”.  
Në datasheet, $V_{th}$ zakonisht matet për rrymë shumë të vogël, p.sh. $I_D = 250\,\mu\text{A}$.

Për switching duhet kontrolluar **$R_{DS(on)}$** te tensioni real i komandimit, p.sh. $V_{GS}=3.3\,\text{V}$, $5\,\text{V}$ ose $10\,\text{V}$.

</div>


# **IV — Procedura Praktike për Analizën e Qarqeve me Transistorë**

<div style="border:3px solid #268; background-color:#eef8ff; color:#023; border-radius: 12px; padding: 12px;">

## Procedurë praktike për analizën e qarqeve me transistorë

1. Hiq sinjalin AC dhe analizo qarkun DC.
2. Gjej pikën e qetësisë:
   * $I_C$, $V_{CE}$ për BJT;
   * $I_D$, $V_{DS}$ për FET/MOSFET.
3. Kontrollo regjimin e punës.
4. Zëvendëso transistorin me modelin përkatës të sinjalit të vogël.
5. Llogarit amplifikimin, rezistencën hyrëse dhe rezistencën dalëse.
6. Kontrollo kufijtë e sinjalit, clipping-un dhe disipimin e fuqisë.
7. Krahaso me simulim SPICE dhe/ose matje laboratorike.

</div>

## **1. Ndarja DC/AC**

Në analizën praktike përdoret shpesh ndarja:

$$
v(t) = V_{DC} + v_{ac}(t)
$$

Për shembull:

$$
v_{BE}(t) = V_{BE,Q} + v_{be}(t)
$$

dhe

$$
i_C(t) = I_{C,Q} + i_c(t)
$$

Q-point përcakton parametrat e modelit të sinjalit të vogël:

Për BJT:

$$
g_m = \frac{I_C}{V_T}
$$

$$
r_\pi = \frac{\beta}{g_m}
$$

Për MOSFET në saturation:

$$
g_m \approx \frac{2I_D}{V_{GS}-V_{th}}
$$

ose

$$
g_m \approx \sqrt{2kI_D}
$$

Këto madhësi nuk janë konstante universale; ato varen nga Q-point.


# **V — Aplikime Praktike**

## **1. BJT si ndërprerës për LED**

**Problemi:** Një LED kërkon $10\,\text{mA}$ për të ndriçuar. Tensioni përçues i LED-it është rreth $2\,\text{V}$. Përdor një BJT NPN për të ndezur LED-in kur aplikohet një sinjal $V_{in}=5\,\text{V}$.

Të dhënat:

* $V_{CC} = 5\,\text{V}$
* $I_C = 10\,\text{mA}$
* $V_{LED} \approx 2\,\text{V}$
* $V_{CE(sat)} \approx 0.2\,\text{V}$
* emiteri në masë;
* LED dhe rezistenca në kolektor.

### **Rezistenca e LED-it**

$$
V_R = V_{CC} - V_{LED} - V_{CE(sat)}
$$

$$
V_R = 5 - 2 - 0.2 = 2.8\,\text{V}
$$

$$
R = \frac{V_R}{I_C}
$$

$$
R = \frac{2.8\,\text{V}}{10\,\text{mA}} = 280\,\Omega
$$

Në praktikë mund të zgjidhet vlera standarde:

$$
R = 270\,\Omega \quad \text{ose} \quad 330\,\Omega
$$

### **Rezistenca e bazës**

Për switching nuk përdorim zakonisht $\beta$ nominale, sepse duam që BJT të hyjë qartë në saturim.  
Përdorim një **forced beta** më të vogël, p.sh.

$$
\beta_{forced} = 10
$$

Prandaj:

$$
I_B = \frac{I_C}{\beta_{forced}}
$$

$$
I_B = \frac{10\,\text{mA}}{10} = 1\,\text{mA}
$$

$$
R_B = \frac{V_{in} - V_{BE}}{I_B}
$$

$$
R_B = \frac{5 - 0.7}{1\,\text{mA}} = 4.3\,\text{k}\Omega
$$

Në praktikë zgjidhet:

$$
R_B \approx 4.7\,\text{k}\Omega
$$

<div style="border:3px solid #b55; background-color:#fff3f0; color:#300; border-radius: 12px; padding: 12px;">

#### Kujdes

Nëse përdoret $\beta=100$ për llogaritjen e rrymës së bazës, del $R_B \approx 43\,\text{k}\Omega$.  
Kjo mund të mos e fusë transistorin në saturim të sigurt, sepse $\beta$ ndryshon shumë dhe bie në kushte reale switching-u.  
Për çelësa me BJT përdoret zakonisht **forced beta**.

</div>


## **2. MOSFET si ndërprerës**

Për switching me MOSFET, kriteri kryesor nuk është $V_{th}$, por:

$$
R_{DS(on)}
$$

te tensioni real i komandimit.

Nëse një MOSFET komandohet nga një mikrokontrollor me $3.3\,\text{V}$, duhet zgjedhur një MOSFET që ka $R_{DS(on)}$ të specifikuar për:

$$
V_{GS}=2.5\,\text{V}
$$

ose

$$
V_{GS}=3.3\,\text{V}
$$

Nëse datasheet jep $R_{DS(on)}$ vetëm për $V_{GS}=10\,\text{V}$, ai MOSFET mund të mos jetë i përshtatshëm për komandim direkt nga mikrokontrollori.

Fuqia e disipuar në MOSFET në gjendjen ON është:

$$
P_{MOSFET} = I_D^2 R_{DS(on)}
$$

Kjo duhet krahasuar me kufijtë termikë të paketës.


# **VI — Modele të Avancuara dhe Simulimi SPICE**

## **1. Pse nevojiten modele të avancuara?**

Modelet e thjeshta si Ebers–Moll, hybrid-$\pi$ ose modeli katror i MOSFET-it janë shumë të dobishme për të kuptuar qarqet.

Por për simulime reale ato nuk mjaftojnë gjithmonë, sepse transistorët realë kanë:

* varësi nga temperatura;
* kapacitanca parazitare;
* efekte të frekuencave të larta;
* efekte të kanalit të shkurtër;
* jolinearitete në rryma të mëdha;
* toleranca prodhimi.

## **2. Modele të avancuara për BJT**

| Modeli | Përdorimi kryesor |
|---|---|
| Gummel–Poon | Model praktik SPICE për BJT me saktësi më të mirë se Ebers–Moll |
| VBIC | Model më i avancuar për BJT në qarqe të integruara |
| HICUM | Model për rryma të mëdha dhe frekuenca shumë të larta |

## **3. Modele të avancuara për MOSFET**

| Modeli | Përdorimi kryesor |
|---|---|
| BSIM3/BSIM4/BSIM6 | Modele standarde për CMOS |
| EKV | Shumë i dobishëm për dizajn analog dhe regjim nënprag |
| HiSIM | Model fizik për teknologji moderne |

## **4. Shembuj sintakse SPICE**

### BJT

```spice
Q1 C B E QNPN
.model QNPN NPN (IS=1e-15 BF=100 VAF=100)
```

### MOSFET

```spice
M1 D G S 0 NMOS
.model NMOS NMOS (VTO=1 KP=50u LAMBDA=0.01)
```

## **5. Analizat kryesore në SPICE**

| Analiza | Qëllimi |
|---|---|
| `.op` | Gjen pikën e punës DC |
| `.dc` | Jep karakteristikat statike $I$–$V$ |
| `.ac` | Analizon fitimin dhe fazën për sinjal të vogël |
| `.tran` | Analizon përgjigjen në kohë |
| `.step` | Ndryshon parametrat për të parë ndjeshmërinë |

# **VII — Përdorim Praktik dhe Ushtrime Laboratorike me Transistorë**

## **1. Matja e karakteristikave $I$–$V$ të BJT**

### BJT — Karakteristika $I_C$ kundrejt $V_{CE}$ për $I_B$ të ndryshme

#### Pajisjet

* BJT NPN, p.sh. 2N3904 ose BC547
* Rezistencë kufizuese për bazën
* Rezistencë ngarkese në kolektor
* Burim DC
* Multimetër ose sensor rryme
* Oshiloskop, nëse kërkohet vëzhgim dinamik

#### Procedura

1. Fikso $I_B$ në një vlerë të vogël, p.sh. $10\,\mu\text{A}$.
2. Ndrysho $V_{CE}$ gradualisht.
3. Mat $I_C$ për çdo vlerë të $V_{CE}$.
4. Përsërit për disa vlera të $I_B$.
5. Vizato familjen e kurbave $I_C(V_{CE})$.

#### Rezultati

Merren kurba që tregojnë:

* cutoff;
* saturim;
* regjimin aktiv;
* efektin Early nëse matjet janë mjaft të sakta.

## **2. Matja e karakteristikave $I$–$V$ të MOSFET**

### MOSFET — Karakteristika $I_D$ kundrejt $V_{DS}$ për $V_{GS}$ të ndryshme

#### Pajisjet

* MOSFET n-channel
* Burim DC për $V_{GS}$
* Burim DC për $V_{DS}$
* Rezistencë kufizuese ose ngarkesë
* Multimetër/ampermetër

#### Procedura

1. Fikso $V_{GS}$ mbi $V_{th}$, p.sh. $2.5\,\text{V}$, $3.3\,\text{V}$, $4\,\text{V}$.
2. Ndrysho $V_{DS}$ gradualisht.
3. Mat $I_D$.
4. Përsërit për çdo vlerë të $V_{GS}$.
5. Vizato familjen e kurbave $I_D(V_{DS})$.

#### Rezultati

Kurbat tregojnë:

* regjimin triode/ohmik;
* regjimin saturation të MOSFET-it;
* varësinë e rrymës nga $V_{GS}$;
* efektin e channel-length modulation nëse vërehet pjerrësi në zonën saturation.


## **3. Testimi i përforcuesve**

### Përforcuesi me emiter të përbashkët

1. Ndërto qarkun me ndarës tensioni në bazë dhe rezistencë në emiter.
2. Mat Q-point: $V_B$, $V_E$, $V_C$, $V_{CE}$.
3. Apliko sinjal sinusoidal të vogël në hyrje, p.sh. $10\,\text{mV}$ deri $20\,\text{mV}$.
4. Mat $v_{out}$ në kolektor.
5. Llogarit:

$$
A_v = \frac{v_{out}}{v_{in}}
$$

6. Kontrollo përmbysjen e fazës dhe clipping-un.

### Përforcuesi common-source me MOSFET

1. Vendos polarizim DC në gate.
2. Përdor $R_D$ në drain dhe, nëse duhet, $R_S$ në source.
3. Mat Q-point: $V_G$, $V_S$, $V_D$, $V_{DS}$.
4. Apliko sinjal AC të vogël në gate.
5. Mat sinjalin në drain dhe llogarit fitimin.

| Simulim i qarkut në SPICE | Krahasim me llogaritje dhe matje |
| Source follower ose emitter follower | Matje e impedancave hyrëse/dalëse |
